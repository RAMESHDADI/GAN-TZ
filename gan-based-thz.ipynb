```python
"""
================================================================================
PHYSICS-GUIDED SUPER-RESOLUTION OF ACTIVE THz IMAGES  --  SINGLE-FILE PIPELINE
(data download + leakage-aware split + 12 methods + statistics + figures + report)
================================================================================
HOW TO RUN (Kaggle, GPU T4, Internet ON):
  * Paste this whole file into ONE cell and run it. Then "Save Version".
  * Next session: attach the previous version's OUTPUT as an input and run the
    same cell again. Everything resumes; finished work is never repeated.
  * Repeat until the log prints "ALL STAGES COMPLETE" (about 4-5 sessions).

WHAT HAPPENS EACH RUN (in order, each stage skipped when already done):
  0. Datasets
     (a) Active THz concealed-object dataset (Liang et al., arXiv:2105.03677;
         github.com/LingLIx/THz_Dataset): copied from an attached Kaggle input if
         present, otherwise downloaded from the authors' Google Drive folder.
     (b) THz-TDS dataset with CAD ground truth (Hung & Yang, IRMMW-THz 2019;
         github.com/wtnthu/THz_Data, CC BY-NC-SA 3.0): real measured THz images
         with an independent ground truth -> external, real-reference test.
  1. Session-grouped split (front/back scans share a timestamp) + leakage audit
  2. Classical baselines: bicubic, PSF-aware Richardson-Lucy
  3. Main queue: 10 learned methods x 3 seeds + 2 extra seeds for the PSF ablation
  4. When the queue is complete: YOLO downstream detection, 5-fold CV,
     external THz-TDS evaluation, final statistics, tables, figures,
     NUMBERS_FOR_TEXT.md and FINAL_EVIDENCE_SUMMARY.txt

If a download fails (Google Drive quota), download manually from the links above,
upload as a Kaggle dataset, attach it and re-run: the cell finds it automatically.
================================================================================
"""

# ==============================================================================
# 0. IMPORTS / INSTALLS
# ==============================================================================
import os, sys, re, json, time, glob, math, random, shutil, hashlib, platform, subprocess, importlib, importlib.util, warnings
import importlib.metadata as _md
from pathlib import Path
from collections import defaultdict, Counter
warnings.filterwarnings("ignore")
try:
    sys.stdout.reconfigure(line_buffering=True)          # show prints immediately (Kaggle buffers output otherwise)
except Exception:
    pass


def stage(msg):
    print(f"\n>>> [{time.strftime('%H:%M:%S')}] {msg}", flush=True)


def _pip(*pkgs):
    cons = Path("/kaggle/working/constraints.txt" if os.path.isdir("/kaggle/working") else "constraints.txt")
    pins = []
    for p in ("numpy", "torch", "torchvision"):
        try:
            pins.append(f"{p}=={_md.version(p)}")
        except Exception:
            pass
    cons.write_text("\n".join(pins) + "\n")
    r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-c", str(cons), *pkgs])
    importlib.invalidate_caches()
    return r.returncode == 0


if importlib.util.find_spec("lpips") is None:
    _pip("lpips")
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
from PIL import Image
import matplotlib
import matplotlib.pyplot as plt
from scipy import stats as sps
from scipy.special import j1 as bessel_j1
try:
    get_ipython().run_line_magic("matplotlib", "inline")
except Exception:
    pass

SESSION_START = time.time()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ==============================================================================
# 1. CONFIGURATION (everything the paper must report is here)
# ==============================================================================
CFG = {
    "RESULTS_DIR": "/kaggle/working/thz_sr_q1",
    "PREVIOUS_RESULTS_DIR": None,
    "DATA_ROOT": None,                      # None = auto-detect the image folder under /kaggle/input
    # ---- degradation / physics (state these in the paper) ----
    "SCALE": 4,
    "PSF_SIGMA_U": 2.0, "PSF_SIGMA_V": 2.0, # Gaussian PSF std in HR pixels (calibrated, not measured)
    "PSF_KSIZE": 25,
    "NOISE_SIGMA": 0.01,                    # additive Gaussian sensor noise on [0,1] LR intensities
    # ---- training (identical for every learned method) ----
    "HR_CROP": 128, "BATCH": 16, "ITERS": 15000, "GAN_WARMUP_ITERS": 1500, "EVAL_EVERY": 1500, "VAL_N": 120,
    "LR_G": 1e-4, "LR_D": 1e-4, "BETAS": (0.9, 0.999), "D_REAL_LABEL": 0.9, "BG_REJECT_STD": 0.02, "BG_REJECT_P": 0.75,
    "VGG_PRETRAINED": True,
    # ---- loss weights (paper defaults) ----
    "W_PIX": 1.0, "W_PERC": 0.1, "W_ADV": 0.005, "W_PSF": 0.5, "W_VAR": 2.0,
    # ---- architecture ----
    "RRDB_NB": 8, "RRDB_NF": 64, "RRDB_GC": 32, "EDSR_NB": 16, "SRRESNET_NB": 16,
    "SWIN_DIM": 60, "SWIN_DEPTHS": (6, 6, 6, 6), "SWIN_HEADS": 6, "SWIN_WS": 8,
    "RL_ITERS": 10,
    # ---- dataset hygiene ----
    "REQUIRE_XML": True,                   # use only images that have their own VOC annotation (the official release)
    "XML_SIZE_TOL": 4,                     # image size must match the <size> in its xml (pixels)
    # ---- split / protocol ----
    "SPLIT_SEED": 2026, "TEST_FRAC": 0.20, "VAL_FRAC": 0.10, "CV_FOLDS": 5, "PAIR_WINDOW_S": 600,
    "NEARDUP_HAMMING": 8, "NEARDUP_CORR": 0.995, "NEARDUP_MAX_GROUP_FRAC": 0.02,
    "SEEDS": [42, 123, 2024], "ABLATION_EXTRA_SEEDS": [7, 99],
    # ---- analyses ----
    "SHIFT_SIGMAS": [1.2, 1.6, 2.0, 2.5], "SHIFT_NOISES": [0.0, 0.02, 0.05, 0.1], "SHIFT_N": 150,
    "SPECTRUM_N": 100, "N_QUAL": 6,
    "SESSION_HOURS": 11.5, "DEFAULT_RUN_HOURS": 0.9, "RUNS_PER_SESSION": 40,
    "SAVE_G_FOR": ["ours", "ours_nopsf", "rrdb_l1", "esrgan", "swinir"],
}
C = CFG
R = Path(CFG["RESULTS_DIR"])
DIRS = {k: R / k for k in ["data", "runs", "metrics", "logs", "figures", "tables", "analysis", "sr_png", "qual", "models", "det"]}

# method registry: generator, GAN?, loss weights, PSF kernel used inside the consistency loss
METHODS = {
    "srcnn":       dict(gen="srcnn",    gan=False, w=dict(pix=1.0)),
    "edsr":        dict(gen="edsr",     gan=False, w=dict(pix=1.0)),
    "swinir":      dict(gen="swinir",   gan=False, w=dict(pix=1.0)),
    "rrdb_l1":     dict(gen="rrdb",     gan=False, w=dict(pix=1.0)),
    "rrdb_l1_psf": dict(gen="rrdb",     gan=False, w=dict(pix=1.0, psf=C["W_PSF"])),
    "srgan":       dict(gen="srresnet", gan=True,  w=dict(pix=C["W_PIX"], perc=C["W_PERC"], adv=C["W_ADV"])),
    "esrgan":      dict(gen="rrdb",     gan=True,  w=dict(pix=C["W_PIX"], perc=C["W_PERC"], adv=C["W_ADV"])),
    "ours_nopsf":  dict(gen="rrdb",     gan=True,  w=dict(pix=C["W_PIX"], perc=C["W_PERC"], adv=C["W_ADV"], var=C["W_VAR"])),
    "ours":        dict(gen="rrdb",     gan=True,  w=dict(pix=C["W_PIX"], perc=C["W_PERC"], adv=C["W_ADV"], var=C["W_VAR"], psf=C["W_PSF"])),
    "ours_airy":   dict(gen="rrdb",     gan=True,  w=dict(pix=C["W_PIX"], perc=C["W_PERC"], adv=C["W_ADV"], var=C["W_VAR"], psf=C["W_PSF"]),
                        psf_loss_kernel="airy"),
}
CLASSICAL = ["bicubic", "rl_deconv"]
LABEL = {"bicubic": "Bicubic", "rl_deconv": "Richardson–Lucy (PSF)", "srcnn": "SRCNN", "edsr": "EDSR", "swinir": "SwinIR-light",
         "rrdb_l1": "RRDB-L1 (ESRNet)", "rrdb_l1_psf": "RRDB-L1 + PSF loss", "srgan": "SRGAN", "esrgan": "ESRGAN-style",
         "ours_nopsf": "Ours w/o PSF loss", "ours": "Ours (PSF loss)", "ours_airy": "Ours (Airy PSF in loss)",
         "hr_proxy": "HR proxy", "lr_nearest": "LR (nearest)"}
COLOR = {"bicubic": "#7f7f7f", "rl_deconv": "#bcbd22", "srcnn": "#c5b0d5", "edsr": "#9467bd", "swinir": "#17becf", "rrdb_l1": "#1f77b4",
         "rrdb_l1_psf": "#aec7e8", "srgan": "#ff9896", "esrgan": "#ff7f0e", "ours_nopsf": "#2ca02c", "ours": "#d62728",
         "ours_airy": "#8c564b", "hr_proxy": "#000000", "lr_nearest": "#c7c7c7"}
ORDER = ["bicubic", "rl_deconv", "srcnn", "edsr", "swinir", "rrdb_l1", "rrdb_l1_psf", "srgan", "esrgan", "ours_nopsf", "ours", "ours_airy"]


def build_queue():
    pri = ["ours", "ours_nopsf", "rrdb_l1", "esrgan", "swinir", "edsr", "srcnn", "srgan", "rrdb_l1_psf", "ours_airy"]
    q = [f"{m}_s{s}" for s in C["SEEDS"] for m in pri]
    q += [f"{m}_s{s}" for s in C["ABLATION_EXTRA_SEEDS"] for m in ("ours", "ours_nopsf")]
    return q


def parse_run(rid):
    fold = None
    m = re.match(r"^cv(\d+)_(.+)$", rid)
    if m:
        fold, rid = int(m.group(1)), m.group(2)
    meth, seed = rid.rsplit("_s", 1)
    return meth, int(seed), fold


# ==============================================================================
# 2. PERSISTENCE / UTILITIES
# ==============================================================================
def _jd(o):
    if isinstance(o, (np.bool_,)): return bool(o)
    if isinstance(o, np.integer): return int(o)
    if isinstance(o, np.floating): return float(o)
    if isinstance(o, np.ndarray): return o.tolist()
    return str(o)


def jdump(o, p):
    Path(p).write_text(json.dumps(o, indent=2, default=_jd), encoding="utf-8")


def jload(p):
    p = Path(p)
    return json.loads(p.read_text(encoding="utf-8")) if p.exists() else None


def restore_previous():
    R.mkdir(parents=True, exist_ok=True)
    fresh = not (R / "runs").exists() or not list((R / "runs").glob("*.json"))
    prev = C["PREVIOUS_RESULTS_DIR"]
    if prev is None:
        cands = []
        for d in range(1, 6):
            cands += glob.glob("/kaggle/input/" + "/".join(["*"] * d) + "/thz_sr_q1/RESULTS_MARKER.json")
        if cands:
            prev = os.path.dirname(max(cands, key=lambda m: ((jload(m) or {}).get("updated", ""),
                                                           len(glob.glob(os.path.join(os.path.dirname(m), "runs", "*.json"))))))
            print(f"[restore] {len(cands)} previous output(s); newest: {prev}")
    if prev and fresh and os.path.abspath(prev) != os.path.abspath(R):
        shutil.copytree(prev, R, dirs_exist_ok=True)
        print(f"[restore] copied {prev}")
    for d in DIRS.values():
        d.mkdir(parents=True, exist_ok=True)
    jdump({"updated": time.strftime("%Y-%m-%d %H:%M:%S")}, R / "RESULTS_MARKER.json")


def env_info():
    out = {"python": platform.python_version(), "torch": torch.__version__, "torchvision": torchvision.__version__,
           "cuda": torch.version.cuda, "gpu": torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU"}
    for p in ("numpy", "lpips", "ultralytics", "scipy"):
        try:
            out[p] = _md.version(p)
        except Exception:
            out[p] = None
    return out


def set_seed(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(s)
    torch.backends.cudnn.benchmark = False
    torch.backends.cudnn.deterministic = True


def free_gpu():
    import gc
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def hours_left():
    return C["SESSION_HOURS"] - (time.time() - SESSION_START) / 3600


def stable_int(*parts):
    return int(hashlib.md5("|".join(map(str, parts)).encode()).hexdigest()[:8], 16)


def apply_style():
    plt.rcParams.update({"font.size": 9, "axes.titlesize": 10, "axes.labelsize": 9, "legend.fontsize": 7.5, "figure.dpi": 110,
                         "axes.spines.top": False, "axes.spines.right": False, "axes.grid": True, "grid.alpha": .25})


def save_fig(name):
    plt.tight_layout()
    for ext in ("pdf", "png"):
        plt.savefig(DIRS["figures"] / f"{name}.{ext}", dpi=None if ext == "pdf" else 300, bbox_inches="tight")
    plt.show()
    plt.close("all")


def save_table(df, name, caption=""):
    df.to_csv(DIRS["tables"] / f"{name}.csv", index=False)
    try:
        (DIRS["tables"] / f"{name}.tex").write_text(df.to_latex(index=False, float_format="%.4f", caption=caption,
                                                                 label=f"tab:{name}"), encoding="utf-8")
    except Exception:
        pass


# ==============================================================================
# 3. DATA: discovery, session grouping, perceptual near-duplicates, frozen split, CV folds
# ==============================================================================
IMG_EXT = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")


def _ds_key(p):
    parts = Path(p).parts
    n = 6 if len(parts) > 5 and parts[3] in ("datasets", "competitions", "notebooks", "models") else 4
    return os.path.join(*parts[:n])


def find_dataset():
    """The attached dataset with the most images (all its sub-folders); VOC xml files are taken from the same dataset."""
    if C["DATA_ROOT"] and os.path.isdir(C["DATA_ROOT"]):
        root = C["DATA_ROOT"]
        imgs = [p for p in glob.glob(f"{root}/**/*", recursive=True) if p.lower().endswith(IMG_EXT)]
    else:
        allp = [p for p in glob.glob("/kaggle/input/**/*", recursive=True) if p.lower().endswith(IMG_EXT) and "/thz_sr_q1/" not in p]
        if not allp:
            raise FileNotFoundError("No images under /kaggle/input; attach the THz dataset or set CFG['DATA_ROOT'].")
        root = Counter(_ds_key(p) for p in allp).most_common(1)[0][0]
        imgs = [p for p in allp if _ds_key(p) == root]
    xmls = {Path(p).stem: p for p in glob.glob(f"{root}/**/*.xml", recursive=True)}
    print(f"[data] dataset root={root}: {len(imgs)} images, {len(xmls)} VOC xml files")
    return root, sorted(imgs), xmls


def session_key(stem):
    """Configuration code of a scan: filename without 'front'/'back' tokens and without timestamp-like digit runs."""
    toks = [t for t in re.split(r"[_\-\s\.]+", stem.lower()) if t]
    toks = [t for t in toks if not re.fullmatch(r"(front|back|fr|bk)\d*", t) and not re.fullmatch(r"\d{6,}", t)]
    return "_".join(toks) or stem.lower()


def scan_time(stem):
    """Seconds from a trailing timestamp token (MMDDhhmmss / YYMMDDhhmmss / YYYYMMDDhhmmss / hhmmss), else None."""
    d = [t for t in re.split(r"[_\-\s\.]+", stem) if re.fullmatch(r"\d{6,14}", t)]
    if not d:
        return None
    t = d[-1]
    hh, mm, ss = int(t[-6:-4]), int(t[-4:-2]), int(t[-2:])
    day = int(t[-8:-6]) if len(t) >= 8 else 0
    mon = int(t[-10:-8]) if len(t) >= 10 else 0
    return (((mon * 31 + day) * 24 + hh) * 60 + mm) * 60 + ss


def load_gray(path):
    a = np.asarray(Image.open(path).convert("L"), dtype=np.uint8)
    s = C["SCALE"]
    return a[: a.shape[0] - a.shape[0] % s, : a.shape[1] - a.shape[1] % s]


def dhash(a, n=16):
    im = np.asarray(Image.fromarray(a).resize((n + 1, n), Image.BILINEAR), dtype=np.int16)
    return np.packbits((im[:, 1:] > im[:, :-1]).flatten())


def thumb(a, n=32):
    t = np.asarray(Image.fromarray(a).resize((n, n), Image.BILINEAR), dtype=np.float32).ravel()
    t = t - t.mean()
    return t / (np.linalg.norm(t) + 1e-8)


def parse_voc(xml_path):
    import xml.etree.ElementTree as ET
    root = ET.parse(xml_path).getroot()
    out = []
    for o in root.findall("object"):
        bb = o.find("bndbox")
        if bb is None:
            continue
        out.append({"cls": o.findtext("name", "obj").strip(), "x1": float(bb.findtext("xmin")), "y1": float(bb.findtext("ymin")),
                    "x2": float(bb.findtext("xmax")), "y2": float(bb.findtext("ymax"))})
    return out


def voc_size(xml_path):
    import xml.etree.ElementTree as ET
    try:
        sz = ET.parse(xml_path).getroot().find("size")
        return int(float(sz.findtext("width"))), int(float(sz.findtext("height")))
    except Exception:
        return None


def invalidate_old_index(reason):
    """A frozen split built on the wrong image set is moved aside together with every result computed on it."""
    bad = R / f"superseded_{time.strftime('%Y%m%d_%H%M%S')}"
    bad.mkdir(parents=True, exist_ok=True)
    for k in ("data", "runs", "metrics", "logs", "sr_png", "qual", "analysis", "models", "det", "tables", "figures"):
        if DIRS[k].exists() and any(DIRS[k].iterdir()):
            shutil.move(str(DIRS[k]), str(bad / k))
        DIRS[k].mkdir(parents=True, exist_ok=True)
    print(f"[data] OLD SPLIT DISCARDED ({reason}); it and all results computed on it were moved to {bad}")


def stage_data():
    idx_p = DIRS["data"] / "index.csv"
    if idx_p.exists():
        df = pd.read_csv(idx_p, keep_default_na=False)
        no_xml = int((df.xml.astype(str).str.len() == 0).sum()) if "xml" in df else len(df)
        if C["REQUIRE_XML"] and no_xml > 0:
            invalidate_old_index(f"{no_xml} of {len(df)} images had no VOC annotation (mixed / resized copies)")
        else:
            print(f"[data] frozen index found: {df.split.value_counts().to_dict()} | groups {df.group.nunique()}")
            return df
    root, imgs, xmls = find_dataset()
    if C["REQUIRE_XML"]:
        n0 = len(imgs)
        imgs = [p for p in imgs if Path(p).stem in xmls]
        print(f"[data] REQUIRE_XML: kept {len(imgs)} of {n0} images that have their own VOC annotation "
              f"({n0 - len(imgs)} un-annotated files ignored, e.g. resized copies)", flush=True)
        ok, bad_size = [], 0
        for p in imgs:
            vs = voc_size(xmls[Path(p).stem])
            if vs:
                with Image.open(p) as im_:
                    w_, h_ = im_.size
                if abs(w_ - vs[0]) > C["XML_SIZE_TOL"] or abs(h_ - vs[1]) > C["XML_SIZE_TOL"]:
                    bad_size += 1
                    continue
            ok.append(p)
        if bad_size:
            print(f"[data] {bad_size} images skipped: their size does not match the size recorded in their xml", flush=True)
        imgs = ok
    rows, seen_px, hashes, thumbs = [], {}, [], []
    for n_read, p in enumerate(imgs, 1):
        if n_read % 500 == 0 or n_read == len(imgs):
            print(f"[data] reading and hashing images {n_read}/{len(imgs)}", flush=True)
        a = load_gray(p)
        px = hashlib.md5(a.tobytes()).hexdigest()
        stem = Path(p).stem
        if px in seen_px:                                     # exact duplicate file content: keep one
            continue
        seen_px[px] = stem
        hashes.append(dhash(a)); thumbs.append(thumb(a))
        rows.append({"path": p, "stem": stem, "H": a.shape[0], "W": a.shape[1], "session": session_key(stem), "t": scan_time(stem),
                     "xml": xmls.get(stem, ""), "n_obj": len(parse_voc(xmls[stem])) if stem in xmls else 0})
    df = pd.DataFrame(rows)
    keep = ~df.stem.duplicated()
    df = df[keep].reset_index(drop=True)
    hashes = [h for h, k in zip(hashes, keep) if k]; thumbs = np.stack([t for t, k in zip(thumbs, keep) if k])
    print(f"[data] {len(df)} unique images ({len(imgs) - len(df)} duplicates removed); {df.session.nunique()} configuration codes")
    par = list(range(len(df)))

    def f(x):
        while par[x] != x:
            par[x] = par[par[x]]; x = par[x]
        return x

    def u(a, b):
        a, b = f(a), f(b)
        if a != b:
            par[max(a, b)] = min(a, b)
    # (1) session pairing: same configuration code AND scan times within PAIR_WINDOW_S (front/back of one session);
    #     codes without timestamps are grouped entirely (conservative)
    n_sess = 0
    for code, g in df.groupby("session"):
        if g.t.isna().all():
            ids = list(g.index)
            for i in ids[1:]:
                u(ids[0], i); n_sess += 1
            continue
        g = g.sort_values("t")
        ids, ts = list(g.index), list(g.t)
        for k in range(1, len(ids)):
            if ts[k] is not None and ts[k - 1] is not None and not pd.isna(ts[k]) and abs(ts[k] - ts[k - 1]) <= C["PAIR_WINDOW_S"]:
                u(ids[k - 1], ids[k]); n_sess += 1
    # (2) perceptual near-duplicates: 256-bit dHash distance AND 32x32 correlation (both must agree)
    H = np.stack(hashes)
    nd_links = []
    for i in range(len(H)):
        ham = np.unpackbits(np.bitwise_xor(H[i], H[i + 1:]), axis=1).sum(1)
        for j in np.where(ham <= C["NEARDUP_HAMMING"])[0]:
            j = i + 1 + int(j)
            if float(thumbs[i] @ thumbs[j]) >= C["NEARDUP_CORR"]:
                nd_links.append((i, j))
    par_sess = par.copy()
    for i, j in nd_links:
        u(i, j)
    big = pd.Series([f(i) for i in range(len(df))]).value_counts().max()
    if big > C["NEARDUP_MAX_GROUP_FRAC"] * len(df):
        print(f"[data] WARNING: near-duplicate links would create a group of {big} images "
              f"(> {C['NEARDUP_MAX_GROUP_FRAC']:.0%}); body scans look alike -> near-duplicate links dropped, session pairing kept")
        par[:] = par_sess; nd_links = []
    n_nd = len(nd_links)
    print(f"[data] session links: {n_sess}; perceptual near-duplicate links: {n_nd}")
    df["group"] = [f(i) for i in range(len(df))]
    gs = df.groupby("group").size()
    print(f"[data] leakage groups: {len(gs)} (sizes: max {gs.max()}, mean {gs.mean():.2f}); perceptual near-dup links: {n_nd}")
    # frozen split at group level
    rng = np.random.RandomState(C["SPLIT_SEED"])
    groups = list(gs.index); rng.shuffle(groups)
    n = len(df); split = {}
    nt = nv = 0
    for g in groups:
        if nt < C["TEST_FRAC"] * n:
            split[g] = "test"; nt += gs[g]
        elif nv < C["VAL_FRAC"] * n:
            split[g] = "val"; nv += gs[g]
        else:
            split[g] = "train"
    df["split"] = df.group.map(split)
    # 5 group folds over ALL images (cross-validation)
    order = list(gs.sort_values(ascending=False).index); rng.shuffle(order)
    fs = [0] * C["CV_FOLDS"]; fold = {}
    for g in order:
        k = int(np.argmin(fs)); fold[g] = k; fs[k] += gs[g]
    df["fold"] = df.group.map(fold)
    for a_, b_ in (("train", "val"), ("train", "test"), ("val", "test")):
        ov = set(df[df.split == a_].group) & set(df[df.split == b_].group)
        assert not ov, f"group leakage {a_}-{b_}"
    df.to_csv(idx_p, index=False)
    summ = {"root": root, "n_images": len(df), "n_groups": int(len(gs)), "largest_group": int(gs.max()),
            "split_images": df.split.value_counts().to_dict(), "split_groups": df.groupby("split").group.nunique().to_dict(),
            "cv_fold_sizes": [int(x) for x in fs], "perceptual_near_dup_links": n_nd,
            "leakage_check": "PASS: no session/near-duplicate group shared between train, val and test",
            "session_links": n_sess, "require_xml": C["REQUIRE_XML"],
            "image_sizes": {f"{h}x{w}": int(n) for (h, w), n in df.groupby(["H", "W"]).size().sort_values(ascending=False).head(5).items()}}
    jdump(summ, DIRS["data"] / "data_summary.json")
    print(json.dumps(summ, indent=1, default=_jd))
    return df


IMG_CACHE = {}


def get_img(path):
    if path not in IMG_CACHE:
        IMG_CACHE[path] = load_gray(path)
    return IMG_CACHE[path]


def to_t(a):
    return torch.from_numpy(a.astype(np.float32) / 255.0)[None, None]


# ==============================================================================
# 4. PHYSICS: PSF kernels and the differentiable forward operator T(x) = D_s(h * x)
# ==============================================================================
def gaussian_kernel(su, sv, ks):
    r = np.arange(ks) - ks // 2
    U, V = np.meshgrid(r, r, indexing="ij")
    k = np.exp(-(U ** 2 / (2 * su ** 2) + V ** 2 / (2 * sv ** 2)))
    return (k / k.sum()).astype(np.float32)


def airy_kernel_fwhm_matched(su, sv, ks):
    """Circular-aperture Airy pattern [2 J1(z)/z]^2 with the same FWHM as the Gaussian (FWHM = 0.8436 * first-zero radius)."""
    sigma = math.sqrt(su * sv)
    r0 = 2.3548 * sigma / 0.8436
    r = np.arange(ks) - ks // 2
    U, V = np.meshgrid(r, r, indexing="ij")
    z = 3.8317 * np.sqrt(U ** 2 + V ** 2) / r0
    with np.errstate(divide="ignore", invalid="ignore"):
        k = np.where(z == 0, 1.0, (2 * bessel_j1(z) / z) ** 2)
    return (k / k.sum()).astype(np.float32)


class ForwardOp(nn.Module):
    def __init__(self, kernel, scale):
        super().__init__()
        self.register_buffer("k", torch.from_numpy(kernel)[None, None])
        self.s, self.p = scale, kernel.shape[0] // 2

    def blur(self, x):
        return F.conv2d(F.pad(x, (self.p,) * 4, mode="reflect"), self.k)

    def forward(self, x):
        return F.avg_pool2d(self.blur(x), self.s)

    def adjoint(self, y):
        z = F.interpolate(y, scale_factor=self.s, mode="nearest") / (self.s ** 2)
        return F.conv2d(F.pad(z, (self.p,) * 4, mode="reflect"), torch.flip(self.k, (2, 3)))


def make_op(kind="gaussian", sigma=None, scale=None):
    su = sv = sigma if sigma is not None else None
    su = su if su is not None else C["PSF_SIGMA_U"]; sv = sv if sv is not None else C["PSF_SIGMA_V"]
    k = gaussian_kernel(su, sv, C["PSF_KSIZE"]) if kind == "gaussian" else airy_kernel_fwhm_matched(su, sv, C["PSF_KSIZE"])
    return ForwardOp(k, scale or C["SCALE"]).to(DEVICE)


OP = None


def degrade_fixed(stem, hr_t, sigma=None, noise=None):
    """Deterministic LR for evaluation: noise seeded by (image, condition) -> identical inputs for every method."""
    sigma = C["PSF_SIGMA_U"] if sigma is None else sigma
    noise = C["NOISE_SIGMA"] if noise is None else noise
    op = OP if (sigma == C["PSF_SIGMA_U"] and C["PSF_SIGMA_U"] == C["PSF_SIGMA_V"]) else make_op("gaussian", sigma)
    with torch.no_grad():
        y = op(hr_t.to(DEVICE))
        if noise > 0:
            g = torch.Generator().manual_seed(stable_int(stem, sigma, noise))
            y = y + noise * torch.randn(y.shape, generator=g).to(DEVICE)
    return y.clamp(0, 1)


# ==============================================================================
# 5. MODELS (all predict a residual over bicubic upsampling, output clamped to [0,1])
# ==============================================================================
def bicubic_up(lr, s=None):
    return F.interpolate(lr, scale_factor=s or C["SCALE"], mode="bicubic", align_corners=False)


def upsampler(nf, s):
    layers = []
    if s in (2, 4, 8):
        for _ in range(int(math.log2(s))):
            layers += [nn.Conv2d(nf, nf * 4, 3, 1, 1), nn.PixelShuffle(2), nn.LeakyReLU(0.2, True)]
    elif s == 3:
        layers += [nn.Conv2d(nf, nf * 9, 3, 1, 1), nn.PixelShuffle(3), nn.LeakyReLU(0.2, True)]
    return nn.Sequential(*layers)


class ResidualSR(nn.Module):
    def forward(self, lr):
        return (bicubic_up(lr) + self.body(lr)).clamp(0, 1)


class SRCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(1, 64, 9, 1, 4), nn.ReLU(True), nn.Conv2d(64, 32, 5, 1, 2), nn.ReLU(True), nn.Conv2d(32, 1, 5, 1, 2))

    def forward(self, lr):
        b = bicubic_up(lr)
        return (b + self.net(b)).clamp(0, 1)


class ResBlock(nn.Module):
    def __init__(self, nf, bn=False):
        super().__init__()
        L = [nn.Conv2d(nf, nf, 3, 1, 1)] + ([nn.BatchNorm2d(nf)] if bn else []) + [nn.PReLU() if bn else nn.ReLU(True), nn.Conv2d(nf, nf, 3, 1, 1)] + \
            ([nn.BatchNorm2d(nf)] if bn else [])
        self.b = nn.Sequential(*L)

    def forward(self, x):
        return x + self.b(x)


class EDSR(ResidualSR):
    def __init__(self, nf=64, nb=16, bn=False):
        super().__init__()
        self.head = nn.Conv2d(1, nf, 3, 1, 1)
        self.trunk = nn.Sequential(*[ResBlock(nf, bn) for _ in range(nb)], nn.Conv2d(nf, nf, 3, 1, 1))
        self.up = upsampler(nf, C["SCALE"]); self.tail = nn.Conv2d(nf, 1, 3, 1, 1)
        nn.init.zeros_(self.tail.weight); nn.init.zeros_(self.tail.bias)

    def body(self, lr):
        h = self.head(lr)
        return self.tail(self.up(h + self.trunk(h)))


class DenseBlock(nn.Module):
    def __init__(self, nf, gc):
        super().__init__()
        self.c = nn.ModuleList([nn.Conv2d(nf + i * gc, gc if i < 4 else nf, 3, 1, 1) for i in range(5)])
        self.a = nn.LeakyReLU(0.2, True)

    def forward(self, x):
        f = [x]
        for i, c in enumerate(self.c):
            o = c(torch.cat(f, 1))
            if i < 4:
                f.append(self.a(o))
        return x + 0.2 * o


class RRDB(nn.Module):
    def __init__(self, nf, gc):
        super().__init__()
        self.b = nn.Sequential(DenseBlock(nf, gc), DenseBlock(nf, gc), DenseBlock(nf, gc))

    def forward(self, x):
        return x + 0.2 * self.b(x)


class RRDBNet(ResidualSR):
    def __init__(self, nf=64, nb=8, gc=32):
        super().__init__()
        self.head = nn.Conv2d(1, nf, 3, 1, 1)
        self.trunk = nn.Sequential(*[RRDB(nf, gc) for _ in range(nb)]); self.trunk_conv = nn.Conv2d(nf, nf, 3, 1, 1)
        self.up = upsampler(nf, C["SCALE"]); self.hr = nn.Sequential(nn.Conv2d(nf, nf, 3, 1, 1), nn.LeakyReLU(0.2, True))
        self.tail = nn.Conv2d(nf, 1, 3, 1, 1)
        nn.init.zeros_(self.tail.weight); nn.init.zeros_(self.tail.bias)

    def body(self, lr):
        h = self.head(lr)
        return self.tail(self.hr(self.up(h + self.trunk_conv(self.trunk(h)))))


# ---- SwinIR (light, pixel-shuffle-direct), compact re-implementation of Liang et al. 2021 ----
def win_part(x, ws):
    B, H, W, Cc = x.shape
    return x.view(B, H // ws, ws, W // ws, ws, Cc).permute(0, 1, 3, 2, 4, 5).reshape(-1, ws * ws, Cc)


def win_rev(w, ws, H, W):
    B = w.shape[0] // ((H // ws) * (W // ws))
    return w.view(B, H // ws, W // ws, ws, ws, -1).permute(0, 1, 3, 2, 4, 5).reshape(B, H, W, -1)


class WindowAttention(nn.Module):
    def __init__(self, dim, ws, heads):
        super().__init__()
        self.ws, self.h = ws, heads
        self.scale = (dim // heads) ** -0.5
        self.qkv = nn.Linear(dim, dim * 3); self.proj = nn.Linear(dim, dim)
        self.table = nn.Parameter(torch.zeros((2 * ws - 1) ** 2, heads)); nn.init.trunc_normal_(self.table, std=.02)
        c = torch.stack(torch.meshgrid(torch.arange(ws), torch.arange(ws), indexing="ij")).flatten(1)
        rel = (c[:, :, None] - c[:, None, :]).permute(1, 2, 0) + (ws - 1)
        self.register_buffer("idx", rel[..., 0] * (2 * ws - 1) + rel[..., 1], persistent=False)

    def forward(self, x, mask=None):
        B_, N, Cc = x.shape
        q, k, v = self.qkv(x).reshape(B_, N, 3, self.h, Cc // self.h).permute(2, 0, 3, 1, 4)
        a = (q * self.scale) @ k.transpose(-2, -1) + self.table[self.idx.view(-1)].view(N, N, -1).permute(2, 0, 1)[None]
        if mask is not None:
            nW = mask.shape[0]
            a = a.view(B_ // nW, nW, self.h, N, N) + mask[None, :, None]
            a = a.view(-1, self.h, N, N)
        return self.proj((a.softmax(-1) @ v).transpose(1, 2).reshape(B_, N, Cc))


class SwinLayer(nn.Module):
    def __init__(self, dim, heads, ws, shift):
        super().__init__()
        self.ws, self.shift = ws, shift
        self.n1, self.n2 = nn.LayerNorm(dim), nn.LayerNorm(dim)
        self.attn = WindowAttention(dim, ws, heads)
        self.mlp = nn.Sequential(nn.Linear(dim, 2 * dim), nn.GELU(), nn.Linear(2 * dim, dim))
        self._masks = {}

    def mask(self, H, W, dev):
        key = (H, W, str(dev))
        if key not in self._masks:
            img = torch.zeros(1, H, W, 1, device=dev); cnt = 0
            for hs in (slice(0, -self.ws), slice(-self.ws, -self.shift), slice(-self.shift, None)):
                for wsl in (slice(0, -self.ws), slice(-self.ws, -self.shift), slice(-self.shift, None)):
                    img[:, hs, wsl, :] = cnt; cnt += 1
            mw = win_part(img, self.ws).squeeze(-1)
            m = mw[:, None, :] - mw[:, :, None]
            self._masks[key] = m.masked_fill(m != 0, -100.0).masked_fill(m == 0, 0.0)
        return self._masks[key]

    def forward(self, x, H, W):
        B, L, Cc = x.shape
        h = self.n1(x).view(B, H, W, Cc)
        if self.shift:
            h = torch.roll(h, (-self.shift, -self.shift), (1, 2))
        w = self.attn(win_part(h, self.ws), self.mask(H, W, x.device) if self.shift else None)
        h = win_rev(w, self.ws, H, W)
        if self.shift:
            h = torch.roll(h, (self.shift, self.shift), (1, 2))
        x = x + h.reshape(B, L, Cc)
        return x + self.mlp(self.n2(x))


class RSTB(nn.Module):
    def __init__(self, dim, depth, heads, ws):
        super().__init__()
        self.layers = nn.ModuleList([SwinLayer(dim, heads, ws, 0 if i % 2 == 0 else ws // 2) for i in range(depth)])
        self.conv = nn.Conv2d(dim, dim, 3, 1, 1)

    def forward(self, x, H, W):
        y = x
        for l in self.layers:
            y = l(y, H, W)
        B, L, Cc = y.shape
        y = self.conv(y.transpose(1, 2).reshape(B, Cc, H, W)).flatten(2).transpose(1, 2)
        return x + y


class SwinIRLight(nn.Module):
    def __init__(self, dim=60, depths=(6, 6, 6, 6), heads=6, ws=8):
        super().__init__()
        self.ws = ws
        self.head = nn.Conv2d(1, dim, 3, 1, 1)
        self.layers = nn.ModuleList([RSTB(dim, d, heads, ws) for d in depths])
        self.norm = nn.LayerNorm(dim); self.after = nn.Conv2d(dim, dim, 3, 1, 1)
        self.up = nn.Sequential(nn.Conv2d(dim, C["SCALE"] ** 2, 3, 1, 1), nn.PixelShuffle(C["SCALE"]))
        nn.init.zeros_(self.up[0].weight); nn.init.zeros_(self.up[0].bias)

    def forward(self, lr):
        H0, W0 = lr.shape[-2:]
        ph, pw = (-H0) % self.ws, (-W0) % self.ws
        x = F.pad(lr, (0, pw, 0, ph), mode="reflect") if (ph or pw) else lr
        H, W = x.shape[-2:]
        f = self.head(x)
        t = f.flatten(2).transpose(1, 2)
        for l in self.layers:
            t = l(t, H, W)
        t = self.norm(t).transpose(1, 2).reshape(f.shape)
        res = self.up(self.after(t) + f)[..., : H0 * C["SCALE"], : W0 * C["SCALE"]]
        return (bicubic_up(lr) + res).clamp(0, 1)


class PatchD(nn.Module):
    def __init__(self):
        super().__init__()
        def blk(i, o, s, norm):
            return [nn.Conv2d(i, o, 4, s, 1)] + ([nn.InstanceNorm2d(o)] if norm else []) + [nn.LeakyReLU(0.2, True)]
        self.net = nn.Sequential(*blk(1, 64, 2, False), *blk(64, 128, 2, True), *blk(128, 256, 2, True), *blk(256, 512, 1, True),
                                 nn.Conv2d(512, 1, 4, 1, 1))

    def forward(self, x):
        return self.net(x)


def build_generator(kind):
    return {"srcnn": lambda: SRCNN(), "edsr": lambda: EDSR(64, C["EDSR_NB"]), "srresnet": lambda: EDSR(64, C["SRRESNET_NB"], bn=True),
            "rrdb": lambda: RRDBNet(C["RRDB_NF"], C["RRDB_NB"], C["RRDB_GC"]),
            "swinir": lambda: SwinIRLight(C["SWIN_DIM"], C["SWIN_DEPTHS"], C["SWIN_HEADS"], C["SWIN_WS"])}[kind]().to(DEVICE)


class VGGLoss(nn.Module):
    def __init__(self):
        super().__init__()
        w = torchvision.models.VGG19_Weights.IMAGENET1K_V1 if C["VGG_PRETRAINED"] else None
        self.f = torchvision.models.vgg19(weights=w).features[:18].eval()          # up to relu3_4
        for p in self.f.parameters():
            p.requires_grad_(False)
        self.register_buffer("m", torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1))
        self.register_buffer("s", torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1))

    def forward(self, a, b):
        n = lambda x: (x.repeat(1, 3, 1, 1) - self.m) / self.s
        return F.l1_loss(self.f(n(a)), self.f(n(b)))


# ==============================================================================
# 6. METRICS (identical implementation for every method)
# ==============================================================================
def _gwin(ks=11, s=1.5):
    g = torch.exp(-(torch.arange(ks) - ks // 2).float() ** 2 / (2 * s ** 2)); g = g / g.sum()
    return (g[:, None] * g[None, :])[None, None]


_WIN = None


def ssim_map(x, y):
    """Wang et al. 2004: 11x11 Gaussian window (sigma 1.5), data range 1, 'valid' region."""
    global _WIN
    if _WIN is None or _WIN.device != x.device:
        _WIN = _gwin().to(x.device)
    C1, C2 = 0.01 ** 2, 0.03 ** 2
    mx, my = F.conv2d(x, _WIN), F.conv2d(y, _WIN)
    sxx = F.conv2d(x * x, _WIN) - mx ** 2; syy = F.conv2d(y * y, _WIN) - my ** 2; sxy = F.conv2d(x * y, _WIN) - mx * my
    return ((2 * mx * my + C1) * (2 * sxy + C2)) / ((mx ** 2 + my ** 2 + C1) * (sxx + syy + C2))


def fg_mask(hr):
    """Body/object foreground of the HR proxy: Otsu threshold on a smoothed image, dilated (fallback: full image)."""
    a = hr[0, 0].detach().cpu().numpy()
    from scipy.ndimage import gaussian_filter, binary_dilation
    b = gaussian_filter(a, 2)
    hist, edges = np.histogram(b, 256, (0, 1)); p = hist / max(1, hist.sum()); w = np.cumsum(p); mu = np.cumsum(p * edges[:-1])
    var = (mu[-1] * w - mu) ** 2 / np.maximum(w * (1 - w), 1e-12)
    m = binary_dilation(b > edges[int(np.argmax(var))], iterations=3)
    return m if 0.01 < m.mean() < 0.99 else np.ones_like(m, bool)


_LPIPS = None


def lpips_fn():
    global _LPIPS
    if _LPIPS is None:
        try:
            import lpips
            _LPIPS = lpips.LPIPS(net="alex", verbose=False).to(DEVICE).eval()
        except Exception as e:
            print(f"[metrics] LPIPS unavailable ({e})"); _LPIPS = False
    return _LPIPS


@torch.no_grad()
def image_metrics(sr, hr, lr, mask=None):
    mse = F.mse_loss(sr, hr).item()
    sm = ssim_map(sr, hr)[0, 0]
    out = {"psnr": 10 * math.log10(1 / (mse + 1e-10)), "ssim": sm.mean().item()}
    if mask is not None:
        mt = torch.from_numpy(mask).to(sr.device)
        out["psnr_fg"] = 10 * math.log10(1 / (((sr - hr) ** 2)[0, 0][mt].mean().item() + 1e-10))
        out["ssim_fg"] = sm[mt[5:-5, 5:-5]].mean().item()
        out["fg_frac"] = float(mask.mean())
    lp = lpips_fn()
    if lp:
        out["lpips"] = lp(sr.repeat(1, 3, 1, 1) * 2 - 1, hr.repeat(1, 3, 1, 1) * 2 - 1).item()
    out["dc_error"] = F.l1_loss(OP(sr), lr).item()                            # data consistency with the observed LR
    lap = F.conv2d(sr, torch.tensor([[[[0., 1, 0], [1, -4, 1], [0, 1, 0]]]], device=sr.device))
    out["sharpness"] = lap.var().item()
    return out


def radial_spectrum(img, nb=64):
    a = img[0, 0].detach().cpu().numpy()
    P = np.abs(np.fft.fftshift(np.fft.fft2(a - a.mean()))) ** 2
    h, w = P.shape
    yy, xx = np.indices(P.shape)
    r = np.sqrt(((yy - h / 2) / (h / 2)) ** 2 + ((xx - w / 2) / (w / 2)) ** 2) / math.sqrt(2)
    b = np.minimum((r * nb).astype(int), nb - 1)
    return np.bincount(b.ravel(), P.ravel(), nb) / np.maximum(np.bincount(b.ravel(), minlength=nb), 1)


# ==============================================================================
# 7. CLASSICAL BASELINES
# ==============================================================================
def rl_deconv(lr, iters=None, op=None):
    """Richardson-Lucy super-resolution with the known forward operator (PSF-aware, no learning)."""
    op = op or OP
    x = bicubic_up(lr).clamp(1e-4, 1)
    for _ in range(iters or C["RL_ITERS"]):
        ratio = lr.clamp_min(1e-4) / op(x).clamp_min(1e-4)
        x = (x * op.adjoint(ratio) * (op.s ** 2)).clamp(1e-4, 1)
    return x.clamp(0, 1)


CLASSICAL_FN = {"bicubic": lambda lr: bicubic_up(lr).clamp(0, 1), "rl_deconv": lambda lr: rl_deconv(lr)}


# ==============================================================================
# 8. TRAINING
# ==============================================================================
class CropSampler:
    def __init__(self, paths, seed):
        self.imgs = [get_img(p) for p in paths]
        self.rng = np.random.RandomState(seed)

    def batch(self, n):
        cs, out = C["HR_CROP"], []
        while len(out) < n:
            a = self.imgs[self.rng.randint(len(self.imgs))]
            if a.shape[0] < cs or a.shape[1] < cs:
                continue
            for _ in range(5):
                y, x = self.rng.randint(0, a.shape[0] - cs + 1), self.rng.randint(0, a.shape[1] - cs + 1)
                c = a[y:y + cs, x:x + cs]
                if c.std() / 255 >= C["BG_REJECT_STD"] or self.rng.rand() > C["BG_REJECT_P"]:
                    break
            if self.rng.rand() < .5:
                c = c[:, ::-1]
            out.append(np.ascontiguousarray(c))
        return torch.from_numpy(np.stack(out).astype(np.float32) / 255.0)[:, None].to(DEVICE)


def val_set(df, n=None):
    v = df[df.split == "val"].sort_values("stem")
    if n:
        v = v.iloc[np.linspace(0, len(v) - 1, min(n, len(v))).astype(int)]
    out = []
    for _, r in v.iterrows():
        hr = to_t(get_img(r.path)).to(DEVICE)
        out.append((hr, degrade_fixed(r.stem, hr)))
    return out


@torch.no_grad()
def quick_val(G, vs):
    G.eval(); ps, ss = [], []
    for hr, lr in vs:
        sr = G(lr)
        ps.append(10 * math.log10(1 / (F.mse_loss(sr, hr).item() + 1e-10))); ss.append(ssim_map(sr, hr).mean().item())
    G.train()
    return float(np.mean(ps)), float(np.mean(ss))


_VGG = None


def train(method, seed, train_paths, vs, rid):
    spec = METHODS[method]; w = spec["w"]
    set_seed(seed)
    G = build_generator(spec["gen"]); G.train()
    D = PatchD().to(DEVICE) if spec["gan"] else None
    global _VGG
    if w.get("perc") and _VGG is None:
        _VGG = VGGLoss().to(DEVICE)
    loss_op = make_op(spec.get("psf_loss_kernel", "gaussian")) if w.get("psf") else None
    oG = torch.optim.Adam(G.parameters(), lr=C["LR_G"], betas=C["BETAS"])
    oD = torch.optim.Adam(D.parameters(), lr=C["LR_D"], betas=C["BETAS"]) if D else None
    bce = nn.BCEWithLogitsLoss()
    sampler = CropSampler(train_paths, seed)
    gnoise = torch.Generator(device=DEVICE).manual_seed(seed)
    best, best_it, best_state, curve, logs = -1.0, 0, None, [], []
    t0 = time.time(); acc = defaultdict(float); nacc = 0
    for it in range(1, C["ITERS"] + 1):
        hr = sampler.batch(C["BATCH"])
        with torch.no_grad():
            y = OP(hr)
            lr = (y + C["NOISE_SIGMA"] * torch.randn(y.shape, generator=gnoise, device=DEVICE)).clamp(0, 1)
        adv_on = D is not None and it > C["GAN_WARMUP_ITERS"]
        sr = G(lr)
        if adv_on:
            oD.zero_grad(set_to_none=True)
            pr, pf = D(hr), D(sr.detach())
            ld = bce(pr, torch.full_like(pr, C["D_REAL_LABEL"])) + bce(pf, torch.zeros_like(pf))
            ld.backward(); oD.step(); acc["d"] += ld.item()
        oG.zero_grad(set_to_none=True)
        terms = {"pix": F.l1_loss(sr, hr)}
        if w.get("perc"):
            terms["perc"] = _VGG(sr, hr)
        if w.get("psf"):
            terms["psf"] = F.l1_loss(loss_op(sr), lr)
        if w.get("var"):
            terms["var"] = F.relu(hr.flatten(1).std(1) - sr.flatten(1).std(1)).mean()
        if adv_on and w.get("adv"):
            pf = D(sr); terms["adv"] = bce(pf, torch.ones_like(pf))
        lg = sum(w[k] * v for k, v in terms.items())
        if not torch.isfinite(lg):
            raise RuntimeError(f"non-finite generator loss at iteration {it}")
        lg.backward(); oG.step()
        for k, v in terms.items():
            acc[k] += v.item()
        nacc += 1
        if it % 250 == 0 and it % C["EVAL_EVERY"] != 0:
            el = time.time() - t0
            print(f"[{rid}] it {it}/{C['ITERS']} loss {lg.item():.4f} | {it / el:.1f} it/s | "
                  f"run ETA {(C['ITERS'] - it) / (it / el) / 60:.0f} min | session left {hours_left():.1f} h", flush=True)
        if it % C["EVAL_EVERY"] == 0 or it == C["ITERS"]:
            p, s = quick_val(G, vs)
            row = {"iter": it, "val_psnr": p, "val_ssim": s, "hours": (time.time() - t0) / 3600, **{f"loss_{k}": v / nacc for k, v in acc.items()}}
            curve.append(row); acc = defaultdict(float); nacc = 0
            flag = ""
            eligible = (D is None) or (it > C["GAN_WARMUP_ITERS"])          # GAN models: only checkpoints with adversarial training
            if eligible and s > best:
                best, best_it, best_state = s, it, {k: v.detach().cpu().clone() for k, v in G.state_dict().items()}; flag = " <- best"
            print(f"[{rid}] it {it}/{C['ITERS']} val PSNR {p:.3f} SSIM {s:.4f} ({row['hours']:.2f}h){flag}")
    G.load_state_dict(best_state); G.eval()
    pd.DataFrame(curve).to_csv(DIRS["logs"] / f"{rid}_curve.csv", index=False)
    th = (time.time() - t0) / 3600
    meta = {"best_iter": best_it, "best_val_ssim": best, "val_psnr_at_best": next(r["val_psnr"] for r in curve if r["iter"] == best_it),
            "train_hours": th, "sec_per_iter": th * 3600 / C["ITERS"], "params_M": sum(p.numel() for p in G.parameters()) / 1e6,
            "selection_rule": ("best validation SSIM among checkpoints after the adversarial warm-up" if D is not None
                               else "best validation SSIM (evaluated every EVAL_EVERY iterations)"),
            "selection_rule_version": 2}
    del D, oG, oD
    free_gpu()
    return G, meta


# ==============================================================================
# 9. EVALUATION OF ONE SYSTEM (learned or classical) ON THE FROZEN TEST SET
# ==============================================================================
@torch.no_grad()
def evaluate_system(fn, rid, df_eval, full_extras=True):
    rows, times, spec_sum, nspec = [], [], None, 0
    qual_stems = list(df_eval.sort_values("stem").stem.iloc[np.linspace(0, len(df_eval) - 1, C["N_QUAL"]).astype(int)])
    png_dir = DIRS["sr_png"] / rid
    if full_extras:
        png_dir.mkdir(parents=True, exist_ok=True)
    qual = {}
    for k, (_, r) in enumerate(df_eval.sort_values("stem").iterrows()):
        if (k + 1) % 100 == 0 or k + 1 == len(df_eval):
            print(f"[{rid}] evaluating test images {k + 1}/{len(df_eval)}", flush=True)
        hr = to_t(get_img(r.path)).to(DEVICE); lr = degrade_fixed(r.stem, hr)
        if torch.cuda.is_available():
            torch.cuda.synchronize()
        t0 = time.time(); sr = fn(lr)
        if torch.cuda.is_available():
            torch.cuda.synchronize()
        times.append(time.time() - t0)
        rows.append({"stem": r.stem, **image_metrics(sr, hr, lr, fg_mask(hr))})
        if full_extras:
            Image.fromarray((sr[0, 0].cpu().numpy() * 255).round().astype(np.uint8)).save(png_dir / f"{r.stem}.png")
            if k < C["SPECTRUM_N"]:
                s_ = radial_spectrum(sr); spec_sum = s_ if spec_sum is None else spec_sum + s_; nspec += 1
            if r.stem in qual_stems:
                qual[r.stem] = (sr[0, 0].cpu().numpy() * 255).round().astype(np.uint8)
    pd.DataFrame(rows).to_csv(DIRS["metrics"] / f"{rid}__test.csv", index=False)
    extra = {"infer_ms_per_image": 1000 * float(np.median(times[3:] if len(times) > 5 else times))}
    if full_extras:
        np.save(DIRS["analysis"] / f"spectrum__{rid}.npy", spec_sum / max(1, nspec))
        np.savez_compressed(DIRS["qual"] / f"{rid}.npz", **qual)
        extra["shift_grid"] = shift_grid(fn, df_eval)
    return pd.DataFrame(rows), extra


@torch.no_grad()
def shift_grid(fn, df_eval):
    sub = df_eval.sort_values("stem").iloc[np.linspace(0, len(df_eval) - 1, min(C["SHIFT_N"], len(df_eval))).astype(int)]
    out = []
    print(f"[shift] robustness grid: {len(C['SHIFT_SIGMAS']) * len(C['SHIFT_NOISES'])} conditions x {len(sub)} images", flush=True)
    for sg in C["SHIFT_SIGMAS"]:
        for nz in C["SHIFT_NOISES"]:
            ps, ss = [], []
            for _, r in sub.iterrows():
                hr = to_t(get_img(r.path)).to(DEVICE); sr = fn(degrade_fixed(r.stem, hr, sg, nz))
                ps.append(10 * math.log10(1 / (F.mse_loss(sr, hr).item() + 1e-10))); ss.append(ssim_map(sr, hr).mean().item())
            out.append({"sigma": sg, "noise": nz, "psnr": float(np.mean(ps)), "ssim": float(np.mean(ss))})
    return out


def supersede_warmup_gan_runs():
    """GAN runs whose selected checkpoint predates adversarial training are moved aside and re-queued."""
    bad = []
    for p in sorted(DIRS["runs"].glob("*.json")):
        r = jload(p)
        if r and r.get("status") == "complete" and r.get("kind") in ("learned", "cv") and (r.get("spec") or {}).get("gan") \
                and int(r.get("best_iter", 0)) <= C["GAN_WARMUP_ITERS"]:
            bad.append(r["run_id"])
    if not bad:
        return []
    dst = R / f"superseded_gan_warmup_{time.strftime('%Y%m%d_%H%M%S')}"
    for k in ("runs", "metrics", "logs", "sr_png", "qual", "analysis", "models"):
        (dst / k).mkdir(parents=True, exist_ok=True)
    for rid in bad:
        moves = [(DIRS["runs"] / f"{rid}.json", "runs"), (DIRS["metrics"] / f"{rid}__test.csv", "metrics"),
                 (DIRS["logs"] / f"{rid}_curve.csv", "logs"), (DIRS["sr_png"] / rid, "sr_png"), (DIRS["qual"] / f"{rid}.npz", "qual"),
                 (DIRS["analysis"] / f"spectrum__{rid}.npy", "analysis"), (DIRS["models"] / f"{rid}_G_fp16.pt", "models")]
        for src, k in moves:
            if src.exists():
                shutil.move(str(src), str(dst / k / src.name))
    print(f"[selection] {len(bad)} GAN run(s) had selected a warm-up (pre-adversarial) checkpoint -> moved to {dst.name} "
          f"and re-queued: {bad}", flush=True)
    return bad


def run_done(rid):
    r = jload(DIRS["runs"] / f"{rid}.json")
    return bool(r) and r.get("status") == "complete"


def est_run_hours(rid=None):
    """Per-method estimate from completed runs of the same method (and kind); falls back to the slowest run seen."""
    recs = [r for r in (jload(p) for p in DIRS["runs"].glob("*.json"))
            if r and r.get("kind") in ("learned", "cv") and r.get("wall_hours")]
    if rid is not None:
        meth, seed, fold = parse_run(rid)
        kind = "cv" if fold is not None else "learned"
        same = [r["wall_hours"] for r in recs if r.get("method") == meth and r.get("kind") == kind]
        if not same:
            same = [r["wall_hours"] for r in recs if r.get("method") == meth]
        if same:
            return max(same) * 1.10
    h = [r["wall_hours"] for r in recs]
    return max(h) * 1.15 if h else C["DEFAULT_RUN_HOURS"]


def execute_classical(name):
    rid = f"{name}_s0"
    if run_done(rid):
        return
    t0 = time.time()
    df = INDEX
    print(f"[{rid}] evaluating classical baseline on {int((df.split == 'test').sum())} test images ...", flush=True)
    _, extra = evaluate_system(CLASSICAL_FN[name], rid, df[df.split == "test"])
    jdump({"run_id": rid, "kind": "classical", "method": name, "seed": 0, "status": "complete", "env": env_info(), "cfg": CFG,
           "wall_hours": (time.time() - t0) / 3600, "params_M": 0.0, **extra}, DIRS["runs"] / f"{rid}.json")
    print(f"[{rid}] evaluated")


def execute_run(rid):
    if run_done(rid):
        print(f"[skip] {rid}"); return jload(DIRS["runs"] / f"{rid}.json")
    if hours_left() < est_run_hours(rid):
        print(f"[time] {rid} needs ~{est_run_hours(rid):.2f}h; {hours_left():.2f}h left -> not started (never truncated)")
        return None
    meth, seed, fold = parse_run(rid)
    t0 = time.time()
    df = INDEX
    if fold is None:
        tr_paths, vs, ev = df[df.split == "train"].path.tolist(), VAL, df[df.split == "test"]
    else:                                                # CV: held-out fold; early-stop set = 10% of training groups
        pool = df[df.fold != fold]; g = sorted(pool.group.unique()); rng = np.random.RandomState(100 + fold); rng.shuffle(g)
        inner = set(g[: max(1, int(0.1 * len(g)))])
        tr_paths = pool[~pool.group.isin(inner)].path.tolist()
        vs = val_set(pool[pool.group.isin(inner)].assign(split="val"), C["VAL_N"]); ev = df[df.fold == fold]
    print(f"\n{'=' * 72}\nRUN {rid}: {METHODS[meth]} | train images {len(tr_paths)} | eval images {len(ev)}\n{'=' * 72}")
    G, meta = train(meth, seed, tr_paths, vs, rid)
    full = fold is None and seed == C["SEEDS"][0]
    _, extra = evaluate_system(lambda lr: G(lr), rid, ev, full_extras=full)
    if full and meth in C["SAVE_G_FOR"]:
        torch.save({k: v.half().cpu() for k, v in G.state_dict().items()}, DIRS["models"] / f"{rid}_G_fp16.pt")
    rec = {"run_id": rid, "kind": "learned" if fold is None else "cv", "method": meth, "seed": seed, "fold": fold, "status": "complete",
           "spec": METHODS[meth], "cfg": CFG, "env": env_info(), **meta, **extra, "wall_hours": (time.time() - t0) / 3600}
    jdump(rec, DIRS["runs"] / f"{rid}.json")
    del G
    free_gpu()
    print(f"[{rid}] done in {rec['wall_hours']:.2f}h (best val SSIM {meta['best_val_ssim']:.4f} at it {meta['best_iter']})")
    return rec


def run_queue(max_runs):
    q = build_queue(); pend = [r for r in q if not run_done(r)]
    print(f"[queue] {len(q) - len(pend)}/{len(q)} complete; next: {pend[:4]}")
    n = 0
    for rid in pend:                                        # a run that does not fit is skipped; shorter runs fill the time
        if n >= max_runs:
            break
        if hours_left() < est_run_hours(rid):
            print(f"[time] {rid} needs ~{est_run_hours(rid):.2f}h, {hours_left():.2f}h left -> skipped this session; trying shorter runs")
            continue
        if execute_run(rid) is not None:
            n += 1
    left = [r for r in q if not run_done(r)]
    print(f"[queue] {len(q) - len(left)}/{len(q)} complete." + (" MAIN QUEUE COMPLETE." if not left else f" Next: {left[:3]}"))


# ==============================================================================
# 10. VALIDATION-ONLY PROGRESS REPORT
# ==============================================================================
def records(kind=None):
    rs = [jload(p) for p in sorted(DIRS["runs"].glob("*.json"))]
    return [r for r in rs if r and r.get("status") == "complete" and (kind is None or r["kind"] == kind)]


def progress_report():
    print(f"\n{'#' * 72}\nPROGRESS REPORT -- VALIDATION ONLY (test statistics are produced in Part 3)\n{'#' * 72}")
    rs = records("learned")
    if not rs:
        return
    t = pd.DataFrame([{"run": r["run_id"], "method": r["method"], "seed": r["seed"], "best_iter": r["best_iter"],
                       "val_SSIM": r["best_val_ssim"], "val_PSNR": r["val_psnr_at_best"], "train_h": r["train_hours"],
                       "params_M": r["params_M"]} for r in rs])
    print(t.round(4).to_string(index=False)); save_table(t, "P1_validation_per_run")
    agg = t.groupby("method").agg(n=("seed", "nunique"), SSIM_mean=("val_SSIM", "mean"), SSIM_sd=("val_SSIM", "std"),
                                  PSNR_mean=("val_PSNR", "mean"), PSNR_sd=("val_PSNR", "std")).reindex([m for m in ORDER if m in set(t.method)])
    print("\n[P2] validation mean ± SD over seeds\n" + agg.round(4).to_string()); save_table(agg.reset_index(), "P2_validation_by_method")
    apply_style()
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
    for m in [m for m in ORDER if m in set(t.method)]:
        cs = [pd.read_csv(DIRS["logs"] / f"{r}_curve.csv") for r in t[t.method == m].run if (DIRS["logs"] / f"{r}_curve.csv").exists()]
        if cs:
            L = min(len(c) for c in cs)
            for a, col in zip(ax, ("val_ssim", "val_psnr")):
                Mx = np.stack([c[col].values[:L] for c in cs])
                a.plot(cs[0]["iter"].values[:L], Mx.mean(0), "o-", ms=3, color=COLOR[m], label=f"{LABEL[m]} (n={len(cs)})")
    ax[0].set_ylabel("validation SSIM"); ax[1].set_ylabel("validation PSNR (dB)")
    for a in ax:
        a.set_xlabel("iteration"); a.axvline(C["GAN_WARMUP_ITERS"], ls=":", c="k", lw=.8)
    ax[0].legend(fontsize=6, ncol=2)
    save_fig("V_fig_training_curves")



# ==============================================================================
# 12. QUEUE STATUS / VALIDATION ABLATION PREVIEW
# ==============================================================================


def queue_status():
    rows = []
    for m in [m for m in ORDER if m in METHODS]:
        seeds = C["SEEDS"] + (C["ABLATION_EXTRA_SEEDS"] if m in ("ours", "ours_nopsf") else [])
        rows.append({"method": m, **{f"s{s}": ("done" if run_done(f"{m}_s{s}") else "-") for s in seeds}})
    print("\n[status]\n" + pd.DataFrame(rows).fillna("").to_string(index=False))


def ablation_preview():
    """Validation-only preview of the seed-paired PSF ablation (the confirmatory test is on the test set in Part 3)."""
    rows = []
    for a, b in (("ours", "ours_nopsf"), ("rrdb_l1_psf", "rrdb_l1"), ("ours_airy", "ours")):
        for s in C["SEEDS"] + C["ABLATION_EXTRA_SEEDS"]:
            ra, rb = jload(DIRS["runs"] / f"{a}_s{s}.json"), jload(DIRS["runs"] / f"{b}_s{s}.json")
            if ra and rb:
                rows.append({"pair": f"{a} − {b}", "seed": s, "dSSIM_val": ra["best_val_ssim"] - rb["best_val_ssim"],
                             "dPSNR_val": ra["val_psnr_at_best"] - rb["val_psnr_at_best"]})
    if not rows:
        return
    d = pd.DataFrame(rows)
    print("\n[P3] seed-paired PSF-loss differences on VALIDATION (preview only)\n" + d.round(4).to_string(index=False))
    g = d.groupby("pair").agg(n=("seed", "size"), dSSIM_mean=("dSSIM_val", "mean"), dSSIM_sd=("dSSIM_val", "std"),
                              dPSNR_mean=("dPSNR_val", "mean"), dPSNR_sd=("dPSNR_val", "std")).reset_index()
    print(g.round(4).to_string(index=False))
    save_table(d, "P3_validation_ablation_preview")
    apply_style()
    fig, ax = plt.subplots(1, 2, figsize=(10, 3.6))
    for k, (col, lab) in enumerate((("dSSIM_val", "Δ validation SSIM"), ("dPSNR_val", "Δ validation PSNR (dB)"))):
        for i, (pair, g_) in enumerate(d.groupby("pair", sort=False)):
            ax[k].scatter([i] * len(g_), g_[col], s=30)
            ax[k].hlines(g_[col].mean(), i - .25, i + .25, color="k")
        ax[k].axhline(0, ls="--", c="gray"); ax[k].set_xticks(range(d.pair.nunique())); ax[k].set_xticklabels(d.pair.unique(), fontsize=7)
        ax[k].set_ylabel(lab)
    save_fig("V_fig_ablation_preview")


DO_DETECTION = True
DO_CV = True
FORCE_FINAL = False           # True = preliminary analysis with missing runs (stamped PRELIMINARY)
FINAL_RESERVE_H = 0.75
ALLOW_CPU = False             # never train on CPU (results would not be comparable and runs would not finish)
CPU_PRELIMINARY_ANALYSIS = True   # without a GPU: build PRELIMINARY tables/figures from the finished runs (no training)       # session time kept for the final analysis when cross-validation runs
DET = {"model": "yolov8s.pt", "epochs": 60, "imgsz": 640, "batch": 16, "conf": 0.001, "iou_nms": 0.6, "conf_pr": 0.25, "boot": 1000}
CV_METHODS = ["rrdb_l1", "ours"]       # reduced for GPU quota: PSF ablation is covered by the 5-seed test-set comparison
TOST_MARGIN = {"ssim": 0.005, "ssim_fg": 0.005, "psnr": 0.10, "psnr_fg": 0.10, "lpips": 0.005, "dc_error": 0.0005}
METRICS_MAIN = ["psnr", "ssim", "psnr_fg", "ssim_fg", "lpips", "dc_error"]
HIGHER_BETTER = {"psnr": True, "ssim": True, "psnr_fg": True, "ssim_fg": True, "lpips": False, "dc_error": False, "sharpness": True}
MLAB = {"psnr": "PSNR (dB)", "ssim": "SSIM", "psnr_fg": "PSNR-FG (dB)", "ssim_fg": "SSIM-FG", "lpips": "LPIPS", "dc_error": "‖T(x̂)−y‖₁",
        "sharpness": "sharpness"}


def holm(p):
    it = sorted(p.items(), key=lambda kv: kv[1]); m = len(it); out, run = {}, 0.0
    for r, (k, v) in enumerate(it):
        run = max(run, min(1.0, (m - r) * v)); out[k] = run
    return out


def test_metrics(rid):
    p = DIRS["metrics"] / f"{rid}__test.csv"
    return pd.read_csv(p).sort_values("stem").reset_index(drop=True) if p.exists() else None


def systems_seed42():
    out = {m: f"{m}_s0" for m in CLASSICAL}
    out.update({m: f"{m}_s{C['SEEDS'][0]}" for m in METHODS})
    return {m: r for m, r in out.items() if run_done(r)}


# ==============================================================================
# A. DOWNSTREAM DETECTION
# ==============================================================================
def get_yolo():
    if importlib.util.find_spec("ultralytics") is None and not _pip("ultralytics"):
        print("[det] ultralytics could not be installed -> detection skipped"); return None
    try:
        os.environ.setdefault("YOLO_VERBOSE", "False")
        from ultralytics import YOLO
        return YOLO
    except Exception as e:
        print(f"[det] ultralytics import failed ({e}) -> detection skipped"); return None


def det_classes():
    cs = set()
    for x in INDEX.xml.dropna():
        if isinstance(x, str) and x:
            cs |= {o["cls"] for o in parse_voc(x)}
    return sorted(cs)


def gt_boxes(row, H, W):
    if not isinstance(row.xml, str) or not row.xml:
        return []
    out = []
    for o in parse_voc(row.xml):
        x1, y1, x2, y2 = max(0, o["x1"]), max(0, o["y1"]), min(W, o["x2"]), min(H, o["y2"])
        if x2 - x1 >= 2 and y2 - y1 >= 2:
            out.append((o["cls"], x1, y1, x2, y2))
    return out


def build_yolo_data(classes):
    base = Path("/kaggle/working/yolo_data") if os.path.isdir("/kaggle/working") else DIRS["det"] / "yolo_data"
    if (base / "data.yaml").exists():
        return base / "data.yaml"
    for sp in ("train", "val"):
        (base / "images" / sp).mkdir(parents=True, exist_ok=True); (base / "labels" / sp).mkdir(parents=True, exist_ok=True)
        for _, r in INDEX[INDEX.split == sp].iterrows():
            a = get_img(r.path); H, W = a.shape
            Image.fromarray(a).save(base / "images" / sp / f"{r.stem}.png")
            lines = [f"{classes.index(c)} {(x1 + x2) / 2 / W:.6f} {(y1 + y2) / 2 / H:.6f} {(x2 - x1) / W:.6f} {(y2 - y1) / H:.6f}"
                     for c, x1, y1, x2, y2 in gt_boxes(r, H, W)]
            (base / "labels" / sp / f"{r.stem}.txt").write_text("\n".join(lines))
    (base / "data.yaml").write_text(f"path: {base}\ntrain: images/train\nval: images/val\nnames:\n" +
                                    "".join(f"  {i}: {c}\n" for i, c in enumerate(classes)))
    return base / "data.yaml"


def train_detector(YOLO, classes):
    w = DIRS["det"] / "detector_best.pt"
    if w.exists():
        return w
    yaml = build_yolo_data(classes)
    print(f"[det] training {DET['model']} on HR-proxy train images ({DET['epochs']} epochs) ...")
    m = YOLO(DET["model"])
    res = m.train(data=str(yaml), epochs=DET["epochs"], imgsz=DET["imgsz"], batch=DET["batch"], seed=0, deterministic=True,
                  project=str(DIRS["det"] / "yolo_runs"), name="hr_proxy", exist_ok=True, workers=2, plots=False, verbose=False)
    best = Path(m.trainer.best)
    shutil.copy(best, w)
    return w


def det_versions():
    v = {}
    hr_dir, lr_dir = DIRS["det"] / "versions" / "hr_proxy", DIRS["det"] / "versions" / "lr_nearest"
    if not hr_dir.exists() or not lr_dir.exists():
        hr_dir.mkdir(parents=True, exist_ok=True); lr_dir.mkdir(parents=True, exist_ok=True)
        for _, r in INDEX[INDEX.split == "test"].iterrows():
            a = get_img(r.path); Image.fromarray(a).save(hr_dir / f"{r.stem}.png")
            lr = degrade_fixed(r.stem, to_t(a).to(DEVICE))
            up = F.interpolate(lr, scale_factor=C["SCALE"], mode="nearest")[0, 0].cpu().numpy()
            Image.fromarray((up * 255).round().astype(np.uint8)).save(lr_dir / f"{r.stem}.png")
    v["hr_proxy"], v["lr_nearest"] = hr_dir, lr_dir
    for m, rid in systems_seed42().items():
        d = DIRS["sr_png"] / rid
        if d.exists() and len(list(d.glob("*.png"))) == int((INDEX.split == "test").sum()):
            v[m] = d
    return v


def predict_dir(model, d):
    out = {}
    files = sorted(Path(d).glob("*.png"))
    for k in range(0, len(files), 32):
        res = model.predict([str(f) for f in files[k:k + 32]], imgsz=DET["imgsz"], conf=DET["conf"], iou=DET["iou_nms"],
                            max_det=300, verbose=False)
        for f, r in zip(files[k:k + 32], res):
            b = r.boxes
            out[f.stem] = (b.xyxy.cpu().numpy(), b.conf.cpu().numpy(), b.cls.cpu().numpy().astype(int))
    return out


def _iou(b, G):
    x1 = np.maximum(b[0], G[:, 0]); y1 = np.maximum(b[1], G[:, 1]); x2 = np.minimum(b[2], G[:, 2]); y2 = np.minimum(b[3], G[:, 3])
    inter = np.clip(x2 - x1, 0, None) * np.clip(y2 - y1, 0, None)
    return inter / ((b[2] - b[0]) * (b[3] - b[1]) + (G[:, 2] - G[:, 0]) * (G[:, 3] - G[:, 1]) - inter + 1e-9)


def match_image(pred, gts, ncls, thr):
    """Image-local greedy matching (VOC). Returns per class: list of (score, tp) and #GT."""
    boxes, scores, cls = pred
    out = {c: ([], 0) for c in range(ncls)}
    for c in range(ncls):
        G = np.array([g[1:] for g in gts if g[0] == c], dtype=float).reshape(-1, 4)
        sel = np.where(cls == c)[0]; sel = sel[np.argsort(-scores[sel])]
        used = np.zeros(len(G), bool); recs = []
        for i in sel:
            tp = 0
            if len(G):
                io = _iou(boxes[i], G); io[used] = -1; j = int(np.argmax(io))
                if io[j] >= thr:
                    used[j] = True; tp = 1
            recs.append((float(scores[i]), tp))
        out[c] = (recs, len(G))
    return out


def ap_from(recs, ngt):
    if ngt == 0:
        return np.nan
    if not recs:
        return 0.0
    r = np.array(recs); o = np.argsort(-r[:, 0]); tp = r[o, 1]
    ctp, cfp = np.cumsum(tp), np.cumsum(1 - tp)
    rec = ctp / ngt; prec = ctp / np.maximum(ctp + cfp, 1e-9)
    mrec = np.r_[0, rec, 1]; mpre = np.r_[0, prec, 0]
    for i in range(len(mpre) - 2, -1, -1):
        mpre[i] = max(mpre[i], mpre[i + 1])
    idx = np.where(mrec[1:] != mrec[:-1])[0]
    return float(np.sum((mrec[idx + 1] - mrec[idx]) * mpre[idx + 1]))


def det_eval(preds, gt, ncls, stems):
    thrs = np.round(np.arange(0.5, 0.96, 0.05), 2)
    M = {t: {s: match_image(preds[s], gt[s], ncls, t) for s in stems} for t in thrs}

    def mAP(t, sample):
        aps = []
        for c in range(ncls):
            recs = [x for s in sample for x in M[t][s][c][0]]; ng = sum(M[t][s][c][1] for s in sample)
            aps.append(ap_from(recs, ng))
        return np.nanmean(aps), aps
    m50, per_cls = mAP(0.5, stems)
    m5095 = float(np.mean([mAP(t, stems)[0] for t in thrs]))
    tp = fp = fn = 0
    for s in stems:
        b, sc, cl = preds[s]; keep = sc >= DET["conf_pr"]
        mm = match_image((b[keep], sc[keep], cl[keep]), gt[s], ncls, 0.5)
        for c in range(ncls):
            t_ = sum(x[1] for x in mm[c][0]); tp += t_; fp += len(mm[c][0]) - t_; fn += mm[c][1] - t_
    P, Rr = tp / max(1, tp + fp), tp / max(1, tp + fn)
    return {"mAP50": float(m50), "mAP50_95": m5095, "precision": P, "recall": Rr, "F1": 2 * P * Rr / max(1e-9, P + Rr),
            "per_class_AP50": per_cls}, (lambda sample: mAP(0.5, sample)[0])


DET_PAIRS = [("ours_nopsf", "rrdb_l1"), ("esrgan", "rrdb_l1"), ("srgan", "edsr"), ("ours", "ours_nopsf"), ("ours", "rrdb_l1"),
             ("rrdb_l1_psf", "rrdb_l1"), ("rrdb_l1", "rl_deconv"), ("ours_airy", "ours")]


def det_pairs(base, point):
    """Paired image-bootstrap comparisons of mAP@50 between methods (same resamples for both members of a pair)."""
    out = []
    for a, b in DET_PAIRS:
        if a in base and b in base:
            d = base[a] - base[b]
            out.append({"A": a, "B": b, "delta_mAP50": float(point[a] - point[b]),
                        "ci95": [float(np.nanpercentile(d, 2.5)), float(np.nanpercentile(d, 97.5))],
                        "p_boot": float(max(1.0 / len(d), min(1.0, 2 * min((d <= 0).mean(), (d >= 0).mean()))))})
    return out


def stage_detection():
    YOLO = get_yolo()
    if YOLO is None:
        return None
    classes = det_classes()
    if not classes:
        print("[det] no VOC annotations found -> detection skipped"); return None
    vers = det_versions()
    need = ["bicubic", "ours"]
    if any(n not in vers for n in need):
        print(f"[det] waiting for seed-42 SR outputs ({[n for n in need if n not in vers]} missing)"); return None
    w = train_detector(YOLO, classes)
    model = YOLO(str(w))
    test = INDEX[INDEX.split == "test"].sort_values("stem")
    gt = {}
    for _, r in test.iterrows():
        a = get_img(r.path); gt[r.stem] = [(classes.index(c), *b) for c, *b in gt_boxes(r, *a.shape)]
    stems = list(test.stem)
    cache = DIRS["det"] / "det_results.json"
    done = jload(cache) or {}
    fns = {}
    for name, d in vers.items():
        pr = predict_dir(model, d)
        res, fn = det_eval(pr, gt, len(classes), stems)
        done[name] = res; fns[name] = fn
        print(f"[det] {name:<14} mAP50 {res['mAP50']:.4f} mAP50-95 {res['mAP50_95']:.4f} P {res['precision']:.3f} R {res['recall']:.3f}")
    rng = np.random.default_rng(0)
    samples = [list(rng.choice(stems, len(stems))) for _ in range(DET["boot"])]
    base = {n: np.array([fns[n](s_) for s_ in samples]) for n in fns}
    for n in fns:
        done[n]["mAP50_ci"] = [float(np.nanpercentile(base[n], 2.5)), float(np.nanpercentile(base[n], 97.5))]
        if "bicubic" in base and n != "bicubic":
            dd = base[n] - base["bicubic"]
            done[n]["delta_vs_bicubic"] = float(done[n]["mAP50"] - done["bicubic"]["mAP50"])
            done[n]["delta_vs_bicubic_ci"] = [float(np.nanpercentile(dd, 2.5)), float(np.nanpercentile(dd, 97.5))]
            done[n]["p_boot_vs_bicubic"] = float(min(1, 2 * min((dd <= 0).mean(), (dd >= 0).mean())))
    done["_pairs"] = det_pairs(base, {n: done[n]["mAP50"] for n in fns})
    done["_classes"] = classes; done["_protocol"] = ("YOLO trained once on HR-proxy train images; evaluated on the test set as "
                                                     "reconstructed by each method from the same synthetic LR; VOC AP, 1000 image bootstraps")
    jdump(done, cache)
    return done


# ==============================================================================
# B. CROSS-VALIDATION
# ==============================================================================
def cv_queue():
    return [f"cv{f}_{m}_s{C['SEEDS'][0]}" for f in range(C["CV_FOLDS"]) for m in CV_METHODS]


def run_cv(max_runs=99):
    for f in range(C["CV_FOLDS"]):
        rid = f"cv{f}_bicubic_s0"
        if not run_done(rid):
            _, ex = evaluate_system(CLASSICAL_FN["bicubic"], rid, INDEX[INDEX.fold == f], full_extras=False)
            jdump({"run_id": rid, "kind": "cv", "method": "bicubic", "seed": 0, "fold": f, "status": "complete", **ex}, DIRS["runs"] / f"{rid}.json")
    q = cv_queue(); pend = [r for r in q if not run_done(r)]
    print(f"[cv] {len(q) - len(pend)}/{len(q)} CV runs complete")
    n = 0
    for rid in pend:
        if n >= max_runs:
            break
        if hours_left() < est_run_hours(rid):
            print(f"[time] {rid} needs ~{est_run_hours(rid):.2f}h, {hours_left():.2f}h left -> skipped this session")
            continue
        if execute_run(rid) is not None:
            n += 1
    print("[cv] CROSS-VALIDATION COMPLETE" if all(run_done(r) for r in q) else f"[cv] {sum(run_done(r) for r in q)}/{len(q)} complete")


# ==============================================================================
# C. FINAL ANALYSIS
# ==============================================================================
def hier_boot(A, B, metric, n=2000, seed=0):
    """Seed-and-image hierarchical bootstrap of mean(A_s) - mean(B_s); A, B: lists of per-image arrays (paired seeds)."""
    rng = np.random.default_rng(seed); K, N = len(A), len(A[0])
    obs = float(np.mean([a.mean() - b.mean() for a, b in zip(A, B)]))
    d = np.empty(n)
    for i in range(n):
        ks = rng.integers(0, K, K); ix = rng.integers(0, N, N)
        d[i] = np.mean([A[k][ix].mean() - B[k][ix].mean() for k in ks])
    return obs, float(np.percentile(d, 2.5)), float(np.percentile(d, 97.5))


def tost(d, margin):
    d = np.asarray(d, float); n = len(d)
    if n < 2 or d.std(ddof=1) == 0:
        return np.nan
    se = d.std(ddof=1) / math.sqrt(n)
    p1 = sps.t.sf((d.mean() + margin) / se, n - 1); p2 = sps.t.cdf((d.mean() - margin) / se, n - 1)
    return float(max(p1, p2))


def final_analysis():
    print(f"\n{'#' * 72}\nFINAL ANALYSIS (held-out test set)\n{'#' * 72}")
    q = build_queue(); missing = [r for r in q if not run_done(r)]
    prelim = bool(missing)
    if missing and not FORCE_FINAL:
        print(f"[final] {len(missing)} main runs missing ({missing[:4]} ...). Finish Parts 1-2 first."); return None
    recs = {r["run_id"]: r for r in records() if r["kind"] in ("learned", "classical")}
    vers = {(r.get("env") or {}).get("torch") for r in recs.values()}
    print(f"[env] torch versions across runs: {vers}")
    seeds_of = defaultdict(list)
    for rid, r in recs.items():
        seeds_of[r["method"]].append(r["seed"])
    PER = {rid: test_metrics(rid) for rid in recs}
    n_test = int((INDEX.split == "test").sum())
    apply_style()

    # ---- T1 main results (one row per method, mean ± SD over seeds of per-seed means)
    rows = []
    for m in ORDER:
        rids = [rid for rid, r in recs.items() if r["method"] == m]
        if not rids:
            continue
        per_seed = pd.DataFrame([PER[r][[k for k in METRICS_MAIN if k in PER[r].columns]].mean() for r in rids if PER[r] is not None])
        row = {"Method": LABEL[m], "key": m, "n_seeds": len(rids)}
        for k in METRICS_MAIN:
            if k in per_seed:
                row[f"{MLAB[k]} mean"] = per_seed[k].mean(); row[f"{MLAB[k]} sd"] = per_seed[k].std(ddof=1) if len(per_seed) > 1 else np.nan
        row["Params (M)"] = recs[rids[0]].get("params_M", 0.0); row["ms / image"] = np.mean([recs[r].get("infer_ms_per_image", np.nan) for r in rids])
        rows.append(row)
    T1 = pd.DataFrame(rows)
    print("\n[T1] Test results (n = %d images; mean ± SD over seeds)\n" % n_test + T1.drop(columns="key").round(4).to_string(index=False))
    save_table(T1.drop(columns="key").round(4), "T1_main_results",
               f"Reconstruction quality on the held-out test set (n={n_test} images, session-grouped). Learned methods: mean ± SD over seeds.")

    # ---- T2 per-image paired comparisons (seed 42): vs bicubic and vs ours
    S42 = systems_seed42()
    rows, pv = [], {}
    for ref in ("bicubic", "ours"):
        for m, rid in S42.items():
            if m == ref or ref not in S42:
                continue
            A, B = PER[rid], PER[S42[ref]]
            for k in ("ssim", "psnr", "ssim_fg", "lpips"):
                if k not in A or k not in B:
                    continue
                d = A[k].values - B[k].values
                better = (d > 0) if HIGHER_BETTER[k] else (d < 0)
                try:
                    w = sps.wilcoxon(A[k].values, B[k].values); p = float(w.pvalue)
                except Exception:
                    p = np.nan
                rank = sps.rankdata(np.abs(d)); rbc = float((rank[d > 0].sum() - rank[d < 0].sum()) / rank.sum()) if rank.sum() else 0.0
                rows.append({"_key": (ref, m, k), "comparison": f"{LABEL[m]} vs {LABEL[ref]}", "metric": MLAB[k], "mean_diff": d.mean(), "median_diff": np.median(d),
                             "win_rate_%": 100 * better.mean(), "wilcoxon_p": p, "rank_biserial": rbc, "n": len(d)})
                pv[(ref, m, k)] = p
    T2 = pd.DataFrame(rows)
    if len(T2):
        adj = holm({k: v for k, v in pv.items() if not np.isnan(v)})
        T2["p_holm"] = [adj.get(k_, np.nan) for k_ in T2["_key"]]
        T2 = T2.drop(columns="_key")
        save_table(T2, "T2_per_image_paired_tests", "Per-image paired comparisons (seed 42), Wilcoxon signed-rank, Holm-adjusted")
        print("\n[T2] per-image tests vs bicubic (SSIM)\n" + T2[T2.comparison.str.endswith("Bicubic") & (T2.metric == "SSIM")].round(5).to_string(index=False))

    # ---- T3 seed-paired ablations
    pairs = [("ours", "ours_nopsf", "PSF-consistency loss (GAN)"), ("rrdb_l1_psf", "rrdb_l1", "PSF-consistency loss (no GAN)"),
             ("ours_airy", "ours", "Airy vs Gaussian PSF in the loss"), ("ours_nopsf", "esrgan", "variance-preservation loss"),
             ("esrgan", "rrdb_l1", "adversarial + perceptual training")]
    rows, abl_data = [], {}
    for a, b, what in pairs:
        ss = sorted(set(seeds_of[a]) & set(seeds_of[b]))
        if len(ss) < 2:
            continue
        for k in ("ssim", "psnr", "ssim_fg", "lpips", "dc_error"):
            if k not in PER[f"{a}_s{ss[0]}"]:
                continue
            A = [PER[f"{a}_s{s}"][k].values for s in ss]; B = [PER[f"{b}_s{s}"][k].values for s in ss]
            d = np.array([x.mean() - y.mean() for x, y in zip(A, B)])
            tt = sps.ttest_rel([x.mean() for x in A], [y.mean() for y in B])
            ci = sps.t.interval(0.95, len(d) - 1, d.mean(), d.std(ddof=1) / math.sqrt(len(d))) if d.std(ddof=1) > 0 else (d.mean(), d.mean())
            obs, lo, hi = hier_boot(A, B, k, n=1000)
            rows.append({"effect": what, "A": LABEL[a], "B": LABEL[b], "metric": MLAB[k], "n_seeds": len(ss), "mean_diff": d.mean(),
                         "sd_diff": d.std(ddof=1), "t_ci95": f"[{ci[0]:.4f}, {ci[1]:.4f}]", "paired_t_p": float(tt.pvalue),
                         "TOST_margin": TOST_MARGIN[k], "TOST_p": tost(d, TOST_MARGIN[k]), "hier_boot_ci95": f"[{lo:.4f}, {hi:.4f}]",
                         "sd_A": float(np.std([x.mean() for x in A], ddof=1)), "sd_B": float(np.std([y.mean() for y in B], ddof=1)),
                         "seeds_A_better": int(((d > 0) if HIGHER_BETTER[k] else (d < 0)).sum())})
            abl_data[(a, b, k)] = (ss, [x.mean() for x in A], [y.mean() for y in B])
    T3 = pd.DataFrame(rows)
    if len(T3):
        print("\n[T3] seed-paired ablations\n" + T3.round(5).to_string(index=False))
        save_table(T3.round(5), "T3_ablations",
                   "Seed-paired ablations: paired t-test, 95% CI, TOST equivalence test (margin column), hierarchical bootstrap")

    # ---- T4 cross-validation
    T4 = None
    cv_rows = []
    for f in range(C["CV_FOLDS"]):
        for m in ["bicubic"] + CV_METHODS:
            rid = f"cv{f}_{m}_s0" if m == "bicubic" else f"cv{f}_{m}_s{C['SEEDS'][0]}"
            t = test_metrics(rid)
            if t is not None:
                cv_rows.append({"fold": f, "method": m, "n": len(t), **{k: t[k].mean() for k in ("psnr", "ssim", "ssim_fg", "lpips") if k in t}})
    if cv_rows:
        cvd = pd.DataFrame(cv_rows)
        T4 = cvd.groupby("method").agg(folds=("fold", "nunique"), images=("n", "sum"), PSNR_mean=("psnr", "mean"), PSNR_sd=("psnr", "std"),
                                       SSIM_mean=("ssim", "mean"), SSIM_sd=("ssim", "std"), SSIM_FG_mean=("ssim_fg", "mean"),
                                       SSIM_FG_sd=("ssim_fg", "std")).reindex([m for m in ["bicubic"] + CV_METHODS if m in set(cvd.method)]).reset_index()
        T4["method"] = T4.method.map(LABEL)
        print("\n[T4] 5-fold cross-validation (mean ± SD across folds)\n" + T4.round(4).to_string(index=False))
        save_table(T4.round(4), "T4_cross_validation", "Session-grouped 5-fold cross-validation (mean ± SD across folds)")
        cvd.to_csv(DIRS["tables"] / "T4b_cv_per_fold.csv", index=False)

    # ---- T5 distribution shift, T7 complexity
    rows = []
    for m, rid in S42.items():
        for g in recs[rid].get("shift_grid", []):
            rows.append({"method": LABEL[m], **g})
    T5 = pd.DataFrame(rows)
    if len(T5):
        save_table(T5.round(4), "T5_distribution_shift", "Test-time PSF width / noise mismatch (no retraining), seed 42")
    T7 = pd.DataFrame([{"method": LABEL[m], "params_M": recs[r].get("params_M", 0), "ms_per_image": recs[r].get("infer_ms_per_image"),
                        "train_h": recs[r].get("train_hours", 0), "sec_per_iter": recs[r].get("sec_per_iter")} for m, r in S42.items()])
    save_table(T7.round(3), "T7_complexity")

    # ---- T6 detection
    det = jload(DIRS["det"] / "det_results.json")
    T6 = None
    if det:
        names = [n for n in ["hr_proxy", "lr_nearest"] + ORDER if n in det]
        T6 = pd.DataFrame([{"input to detector": LABEL[n], "mAP@50": det[n]["mAP50"], "mAP@50 95% CI": "[%.3f, %.3f]" % tuple(det[n]["mAP50_ci"]),
                            "mAP@50:95": det[n]["mAP50_95"], "precision": det[n]["precision"], "recall": det[n]["recall"], "F1": det[n]["F1"],
                            "ΔmAP@50 vs bicubic": det[n].get("delta_vs_bicubic", np.nan),
                            "Δ 95% CI": "[%.3f, %.3f]" % tuple(det[n]["delta_vs_bicubic_ci"]) if "delta_vs_bicubic_ci" in det[n] else ""}
                           for n in names])
        print("\n[T6] downstream detection\n" + T6.round(4).to_string(index=False))
        save_table(T6.round(4), "T6_detection", "Concealed-object detection (YOLO trained on HR-proxy images) on reconstructed test images")
        if det.get("_pairs"):
            T6c = pd.DataFrame([{"A": LABEL[x["A"]], "B": LABEL[x["B"]], "ΔmAP@50 (A−B)": x["delta_mAP50"],
                                 "95% CI": "[%.3f, %.3f]" % tuple(x["ci95"]), "p (paired bootstrap)": x["p_boot"]} for x in det["_pairs"]])
            print("\n[T6c] paired detection comparisons (seed 42, 1000 image bootstraps)\n" + T6c.round(4).to_string(index=False))
            save_table(T6c.round(4), "T6c_detection_paired", "Paired image-bootstrap comparisons of detection mAP@50 between methods (seed 42)")
        pcd = pd.DataFrame({LABEL[n]: det[n]["per_class_AP50"] for n in names}, index=det["_classes"]).T
        pcd.round(4).reset_index().rename(columns={"index": "input"}).to_csv(DIRS["tables"] / "T6b_detection_per_class_AP50.csv", index=False)

    T8 = band_error_analysis(S42)
    ctx = dict(recs=recs, PER=PER, S42=S42, T1=T1, T2=T2, T3=T3, T4=T4, T5=T5, T6=T6, det=det, abl=abl_data, prelim=prelim, n_test=n_test, T8=T8,
               seeds_of=seeds_of)
    paper_figures(ctx)
    numbers_for_text(ctx)
    evidence_summary(ctx)
    assets_index()
    return ctx


def band_error_analysis(S42):
    """Split each reconstruction error into the band the measurement can observe (below the LR Nyquist frequency,
    i.e. the range of the forward operator up to PSF attenuation) and the band it cannot (above it, null space of
    down-sampling). A data-consistency / PSF loss can only constrain the first band."""
    cache = DIRS["analysis"] / "band_errors.csv"
    test = INDEX[INDEX.split == "test"].sort_values("stem")
    rows = []
    done = pd.read_csv(cache) if cache.exists() else None
    for m, rid in S42.items():
        if done is not None and m in set(done.method):
            rows += done[done.method == m].to_dict("records"); continue
        d = DIRS["sr_png"] / rid
        if not d.exists():
            continue
        acc = []
        for _, r in test.iterrows():
            f = d / f"{r.stem}.png"
            if not f.exists():
                continue
            hr = get_img(r.path).astype(np.float64) / 255.0
            sr = np.asarray(Image.open(f), dtype=np.float64) / 255.0
            if sr.shape != hr.shape:
                continue
            D = np.fft.fft2(sr - hr); Hh, Ww = D.shape
            low = (np.abs(np.fft.fftfreq(Hh))[:, None] <= 0.5 / C["SCALE"]) & (np.abs(np.fft.fftfreq(Ww))[None, :] <= 0.5 / C["SCALE"])
            P = np.abs(D) ** 2 / (Hh * Ww) ** 2
            S = np.abs(np.fft.fft2(sr - sr.mean())) ** 2 / (Hh * Ww) ** 2
            Hs = np.abs(np.fft.fft2(hr - hr.mean())) ** 2 / (Hh * Ww) ** 2
            acc.append((P[low].sum(), P[~low].sum(), S[~low].sum(), Hs[~low].sum()))
        if not acc:
            continue
        a = np.array(acc)
        rows.append({"method": m, "n": len(a), "rmse_observable": float(np.sqrt(a[:, 0].mean())),
                     "rmse_unobservable": float(np.sqrt(a[:, 1].mean())),
                     "unobservable_share_of_error_%": float(100 * a[:, 1].sum() / (a[:, 0].sum() + a[:, 1].sum())),
                     "hf_energy_ratio_sr_over_hr": float(a[:, 2].sum() / max(a[:, 3].sum(), 1e-12))})
    if not rows:
        return None
    T8 = pd.DataFrame(rows)
    T8.to_csv(cache, index=False)
    T8 = T8.set_index("method").reindex([m for m in ORDER if m in set(T8.method)]).reset_index()
    out = T8.copy(); out["method"] = out.method.map(LABEL)
    save_table(out.round(5), "T8_observable_vs_unobservable_error",
               "Reconstruction error split at the LR Nyquist frequency (seed 42): the observable band is constrained by the "
               "measurement (and by any PSF/data-consistency loss); the unobservable band is not. HF energy ratio > 1 = "
               "more high-frequency content than the HR proxy (synthesised detail).")
    print("\n[T8] observable vs unobservable error (seed 42)\n" + out.round(5).to_string(index=False))
    return T8


# ==============================================================================
# FIGURES
# ==============================================================================
CAP = {}


def _safe(fn):
    def w(*a, **k):
        try:
            return fn(*a, **k)
        except Exception as e:
            print(f"[fig] {fn.__name__} skipped: {type(e).__name__}: {e}")
    w.__name__ = fn.__name__
    return w


@_safe
def fig01_physics(ctx):
    kg = gaussian_kernel(C["PSF_SIGMA_U"], C["PSF_SIGMA_V"], C["PSF_KSIZE"]); ka = airy_kernel_fwhm_matched(C["PSF_SIGMA_U"], C["PSF_SIGMA_V"], C["PSF_KSIZE"])
    fig, ax = plt.subplots(1, 4, figsize=(14, 3.2))
    ax[0].imshow(kg, cmap="inferno"); ax[0].set_title(f"(a) Gaussian PSF σ={C['PSF_SIGMA_U']} px"); ax[0].axis("off")
    ax[1].imshow(np.log10(ka + 1e-6), cmap="inferno"); ax[1].set_title("(b) Airy PSF (log), FWHM-matched"); ax[1].axis("off")
    c = C["PSF_KSIZE"] // 2
    ax[2].plot(np.arange(-c, c + 1), kg[c] / kg.max(), label="Gaussian"); ax[2].plot(np.arange(-c, c + 1), ka[c] / ka.max(), label="Airy")
    ax[2].set_xlabel("HR pixels"); ax[2].set_title("(c) central profiles"); ax[2].legend()
    for k, lab in ((kg, "Gaussian"), (ka, "Airy")):
        Kf = np.abs(np.fft.fftshift(np.fft.fft2(k, (256, 256))))
        ax[3].plot(np.linspace(0, 1, 128), Kf[128, 128:] / Kf.max(), label=f"MTF {lab}")
    ax[3].axvline(1 / C["SCALE"], ls="--", c="k", lw=.8, label="LR Nyquist (1/s)")
    ax[3].set_xlabel("normalised spatial frequency (HR Nyquist = 1)"); ax[3].set_title("(d) modulation transfer"); ax[3].legend(fontsize=6.5)
    save_fig("FIG01_physics_forward_model")
    CAP["FIG01_physics_forward_model"] = (f"Forward model y = D_s(h ⊛ x) + n with s={C['SCALE']}, noise σ={C['NOISE_SIGMA']}. "
                                          "The PSF width is a calibrated value consistent with the dataset's stated ≈5 mm resolution, not a measured PSF.")


def _qual_images(ctx, methods):
    stems = None; arrs = {}
    for m in methods:
        rid = ctx["S42"].get(m)
        f = DIRS["qual"] / f"{rid}.npz" if rid else None
        if f and f.exists():
            z = np.load(f); arrs[m] = {k: z[k] for k in z.files}; stems = sorted(z.files) if stems is None else stems
    return stems or [], arrs


@_safe
def fig02_qualitative(ctx):
    methods = [m for m in ["bicubic", "rl_deconv", "swinir", "rrdb_l1", "esrgan", "ours_nopsf", "ours"] if m in ctx["S42"]]
    stems, arrs = _qual_images(ctx, methods)
    stems = stems[:3]
    if not stems:
        return
    cols = ["lr_nearest"] + methods + ["hr_proxy"]
    fig, ax = plt.subplots(2 * len(stems), len(cols), figsize=(1.9 * len(cols), 3.0 * len(stems)))
    for i, st in enumerate(stems):
        row = INDEX[INDEX.stem == st].iloc[0]; hr = get_img(row.path)
        lr = degrade_fixed(st, to_t(hr).to(DEVICE))
        lrn = (F.interpolate(lr, scale_factor=C["SCALE"], mode="nearest")[0, 0].cpu().numpy() * 255).astype(np.uint8)
        H, W = hr.shape
        boxes = gt_boxes(row, H, W)
        if boxes:
            _, x1, y1, x2, y2 = boxes[0]; cx, cy = (x1 + x2) / 2, (y1 + y2) / 2
        else:
            cy, cx = np.unravel_index(np.argmax(hr), hr.shape)
        h = max(48, int(1.6 * (max(y2 - y1, x2 - x1) if boxes else 48)))
        ys, xs = int(np.clip(cy - h / 2, 0, H - h)), int(np.clip(cx - h / 2, 0, W - h))
        for j, cname in enumerate(cols):
            img = lrn if cname == "lr_nearest" else hr if cname == "hr_proxy" else arrs[cname][st]
            ax[2 * i, j].imshow(img, cmap="gray", vmin=0, vmax=255); ax[2 * i, j].add_patch(plt.Rectangle((xs, ys), h, h, fill=False, ec="y", lw=.8))
            ax[2 * i + 1, j].imshow(img[ys:ys + h, xs:xs + h], cmap="gray", vmin=0, vmax=255)
            if cname not in ("lr_nearest", "hr_proxy"):
                ps = 10 * math.log10(1 / (np.mean((img / 255. - hr / 255.) ** 2) + 1e-10))
                ax[2 * i + 1, j].set_xlabel(f"{ps:.2f} dB", fontsize=6.5)
            if i == 0:
                ax[0, j].set_title(LABEL[cname], fontsize=7)
        for a in ax[2 * i:2 * i + 2].ravel():
            a.set_xticks([]); a.set_yticks([])
    save_fig("FIG02_qualitative")
    CAP["FIG02_qualitative"] = "Test examples (seed 42): full frame and zoom on an annotated concealed object; numbers are image PSNR against the HR proxy."


@_safe
def fig03_main(ctx):
    T1 = ctx["T1"]
    ks = [k for k in ("ssim", "ssim_fg", "psnr", "lpips") if f"{MLAB[k]} mean" in T1]
    fig, ax = plt.subplots(1, len(ks), figsize=(3.8 * len(ks), 3.8))
    for a, k in zip(ax, ks):
        for i, (_, r) in enumerate(T1.iterrows()):
            m = r.key; rids = [rid for rid, rr in ctx["recs"].items() if rr["method"] == m]
            v = [ctx["PER"][rid][k].mean() for rid in rids]
            a.bar(i, np.mean(v), color=COLOR[m], alpha=.6); a.scatter([i] * len(v), v, c="k", s=10, zorder=3)
        a.set_xticks(range(len(T1))); a.set_xticklabels(T1.Method, rotation=55, ha="right", fontsize=6.5)
        allv = T1[f"{MLAB[k]} mean"]; pad = (allv.max() - allv.min()) * .25 + 1e-6
        a.set_ylim(allv.min() - pad, allv.max() + pad); a.set_title(MLAB[k] + (" ↑" if HIGHER_BETTER[k] else " ↓"))
    save_fig("FIG03_main_results")
    CAP["FIG03_main_results"] = "Held-out test results per method: bars = mean over seeds, dots = individual seeds. SSIM-FG/PSNR-FG are computed on the body/object foreground."


@_safe
def fig04_perception_distortion(ctx):
    if "lpips" not in next(iter(ctx["PER"].values())):
        return
    plt.figure(figsize=(6, 4.6))
    for rid, r in ctx["recs"].items():
        t = ctx["PER"][rid]; m = r["method"]
        plt.scatter(t.psnr.mean(), t.lpips.mean(), color=COLOR[m], marker="*" if r["kind"] == "classical" else "o", s=60 if r["kind"] == "classical" else 25,
                    label=LABEL[m] if r["seed"] in (0, C["SEEDS"][0]) else None)
    plt.xlabel("PSNR (dB) ↑"); plt.ylabel("LPIPS ↓"); plt.legend(fontsize=6, ncol=2); plt.title("Perception–distortion plane (each point = one run)")
    save_fig("FIG04_perception_distortion")
    CAP["FIG04_perception_distortion"] = "Perception–distortion trade-off [Blau & Michaeli]: every trained run (circles) and the classical baselines (stars)."


@_safe
def fig05_foreground(ctx):
    S42, PER = ctx["S42"], ctx["PER"]
    ms = [m for m in ORDER if m in S42 and "ssim_fg" in PER[S42[m]]]
    fig, ax = plt.subplots(1, 2, figsize=(12, 3.8))
    x = np.arange(len(ms))
    ax[0].bar(x - .2, [PER[S42[m]].ssim.mean() for m in ms], .4, label="full image", color="#999")
    ax[0].bar(x + .2, [PER[S42[m]].ssim_fg.mean() for m in ms], .4, label="foreground only", color="#d62728")
    ax[0].set_xticks(x); ax[0].set_xticklabels([LABEL[m] for m in ms], rotation=50, ha="right", fontsize=6.5); ax[0].legend(); ax[0].set_ylabel("SSIM")
    lo = min(min(PER[S42[m]].ssim.mean(), PER[S42[m]].ssim_fg.mean()) for m in ms)
    ax[0].set_ylim(max(0, lo - .05), 1); ax[0].set_title("(a) SSIM on full image vs foreground (seed 42)")
    for m in ("bicubic", "ours"):
        if m in S42:
            t = PER[S42[m]]; ax[1].scatter(t.fg_frac, t.ssim - t.ssim_fg, s=6, alpha=.5, color=COLOR[m], label=LABEL[m])
    ax[1].set_xlabel("foreground fraction of the image"); ax[1].set_ylabel("SSIM(full) − SSIM(foreground)"); ax[1].legend()
    ax[1].set_title("(b) full-image minus foreground SSIM vs foreground share")
    save_fig("FIG05_full_vs_foreground")
    CAP["FIG05_full_vs_foreground"] = ("Full-image vs foreground-masked SSIM. Large uniform background regions can dominate full-image SSIM; "
                                       "the foreground-masked SSIM isolates reconstruction of the body and concealed objects.")


@_safe
def fig06_per_image(ctx):
    S42, PER = ctx["S42"], ctx["PER"]
    if not ("ours" in S42 and "bicubic" in S42):
        return
    a, b = PER[S42["ours"]], PER[S42["bicubic"]]
    fig, ax = plt.subplots(1, 3, figsize=(13, 3.8))
    for j, k in enumerate(("ssim", "ssim_fg")):
        ax[j].scatter(b[k], a[k], s=5, alpha=.5, color="#d62728"); lo = min(a[k].min(), b[k].min()); hi = max(a[k].max(), b[k].max())
        ax[j].plot([lo, hi], [lo, hi], "k--", lw=.8)
        ax[j].set_xlabel(f"Bicubic {MLAB[k]}"); ax[j].set_ylabel(f"Ours {MLAB[k]}")
        ax[j].set_title(f"({'ab'[j]}) ours better on {(a[k] > b[k]).mean():.1%} of images")
    d = a.psnr - b.psnr
    ax[2].hist(d, 50, color="#1f77b4"); ax[2].axvline(0, c="k", lw=.8)
    ax[2].set_xlabel("PSNR(ours) − PSNR(bicubic) per image (dB)"); ax[2].set_title(f"(c) PSNR better on {(d > 0).mean():.1%}")
    save_fig("FIG06_per_image_vs_bicubic")
    CAP["FIG06_per_image_vs_bicubic"] = "Per-image comparison of Ours (seed 42) with bicubic on every test image."


@_safe
def fig07_ablation(ctx):
    abl = ctx["abl"]
    panels = [(a, b, k) for (a, b, k) in abl if k in ("ssim", "ssim_fg", "psnr", "dc_error") and (a, b) in (("ours", "ours_nopsf"), ("rrdb_l1_psf", "rrdb_l1"))]
    if not panels:
        return
    ks = [k for k in ("ssim", "ssim_fg", "psnr", "dc_error") if any(p[2] == k for p in panels)]
    fig, ax = plt.subplots(2, len(ks), figsize=(3.5 * len(ks), 6.4), squeeze=False)
    for i, (a, b) in enumerate((("ours", "ours_nopsf"), ("rrdb_l1_psf", "rrdb_l1"))):
        for j, k in enumerate(ks):
            if (a, b, k) not in abl:
                ax[i, j].axis("off"); continue
            ss, A, B = abl[(a, b, k)]
            for s, x, y in zip(ss, A, B):
                ax[i, j].plot([0, 1], [y, x], "o-", color="#d62728" if ((x > y) == HIGHER_BETTER[k]) else "#1f77b4", alpha=.8)
            ax[i, j].set_xticks([0, 1]); ax[i, j].set_xticklabels([LABEL[b], LABEL[a]], fontsize=6.5, rotation=10)
            ax[i, j].set_title(f"{MLAB[k]} (n={len(ss)} seeds)", fontsize=8)
    save_fig("FIG07_psf_ablation_seed_paired")
    CAP["FIG07_psf_ablation_seed_paired"] = ("Seed-paired effect of the PSF-consistency loss with adversarial training (top) and without it (bottom); "
                                             "each line is one seed (red = with-PSF run better).")


@_safe
def fig08_data_consistency(ctx):
    T1 = ctx["T1"]; col = f"{MLAB['dc_error']} mean"
    if col not in T1:
        return
    plt.figure(figsize=(8, 3.6))
    plt.bar(range(len(T1)), T1[col], yerr=T1[f"{MLAB['dc_error']} sd"].fillna(0), color=[COLOR[k] for k in T1.key], capsize=3)
    plt.xticks(range(len(T1)), T1.Method, rotation=50, ha="right", fontsize=6.5)
    plt.ylabel("mean |T(x̂) − y|  (LR intensity units)"); plt.axhline(C["NOISE_SIGMA"] * math.sqrt(2 / math.pi), ls="--", c="k", lw=.8,
                                                                      label="expected |noise| (noise floor)")
    plt.legend(); plt.title("Physical data consistency of the reconstructions")
    save_fig("FIG08_data_consistency")
    CAP["FIG08_data_consistency"] = ("Data-consistency error between the re-degraded reconstruction and the observed LR image; the dashed line is "
                                     "the error expected from sensor noise alone. This is the quantity the PSF-consistency loss optimises.")


@_safe
def fig09_spectrum(ctx):
    plt.figure(figsize=(6.5, 4.4))
    sub = INDEX[INDEX.split == "test"].sort_values("stem").iloc[:C["SPECTRUM_N"]]
    hrs = [radial_spectrum(to_t(get_img(p))) for p in sub.path]
    f = (np.arange(64) + .5) / 64
    plt.semilogy(f, np.mean(hrs, 0), "k-", lw=2, label="HR proxy")
    for m in ("bicubic", "rl_deconv", "rrdb_l1", "esrgan", "ours_nopsf", "ours"):
        rid = ctx["S42"].get(m)
        if rid and (DIRS["analysis"] / f"spectrum__{rid}.npy").exists():
            plt.semilogy(f, np.load(DIRS["analysis"] / f"spectrum__{rid}.npy"), color=COLOR[m], label=LABEL[m])
    plt.axvline(1 / C["SCALE"], ls="--", c="gray", lw=.8, label="LR Nyquist")
    plt.xlabel("normalised radial frequency"); plt.ylabel("power"); plt.legend(fontsize=6.5); plt.title("Radially averaged power spectra (test)")
    save_fig("FIG09_power_spectra")
    CAP["FIG09_power_spectra"] = ("Spectral content of reconstructions vs the HR proxy. Power above the LR Nyquist frequency is synthesised by "
                                  "the model; it can only be compared with the proxy, not with a physically measured sub-diffraction reference.")


@_safe
def fig10_shift(ctx):
    T5 = ctx["T5"]
    if T5 is None or not len(T5):
        return
    ms = [LABEL[m] for m in ("bicubic", "swinir", "rrdb_l1", "ours_nopsf", "ours") if LABEL[m] in set(T5.method)]
    fig, ax = plt.subplots(2, len(ms), figsize=(3 * len(ms), 5.6), squeeze=False)
    for j, m in enumerate(ms):
        for i, k in enumerate(("psnr", "ssim")):
            P = T5[T5.method == m].pivot(index="noise", columns="sigma", values=k)
            im = ax[i, j].imshow(P.values, cmap="viridis", aspect="auto")
            for (y, x), v in np.ndenumerate(P.values):
                ax[i, j].text(x, y, f"{v:.2f}" if k == "psnr" else f"{v:.3f}", ha="center", va="center", fontsize=6, color="w")
            ax[i, j].set_xticks(range(P.shape[1])); ax[i, j].set_xticklabels(P.columns); ax[i, j].set_yticks(range(P.shape[0]))
            ax[i, j].set_yticklabels(P.index); ax[i, j].set_title(f"{m} · {k.upper()}", fontsize=7)
            ax[i, j].set_xlabel("test PSF σ (px)"); ax[i, j].set_ylabel("noise σ")
    save_fig("FIG10_distribution_shift")
    CAP["FIG10_distribution_shift"] = (f"Robustness to a mismatch between the training degradation (σ={C['PSF_SIGMA_U']} px, noise {C['NOISE_SIGMA']}) "
                                       "and test-time PSF width / noise, without retraining (seed 42).")


@_safe
def fig11_detection(ctx):
    det = ctx["det"]
    if not det:
        return
    names = [n for n in ["hr_proxy", "lr_nearest"] + ORDER if n in det]
    fig, ax = plt.subplots(1, 2, figsize=(13, 4))
    v = [det[n]["mAP50"] for n in names]; ci = np.array([det[n]["mAP50_ci"] for n in names])
    ax[0].bar(range(len(names)), v, color=[COLOR[n] for n in names], yerr=[np.array(v) - ci[:, 0], ci[:, 1] - np.array(v)], capsize=3)
    ax[0].set_xticks(range(len(names))); ax[0].set_xticklabels([LABEL[n] for n in names], rotation=50, ha="right", fontsize=6.5)
    ax[0].set_ylabel("mAP@50 (95% bootstrap CI)"); ax[0].set_title("(a) concealed-object detection on reconstructed test images")
    P = np.array([det[n]["per_class_AP50"] for n in names], dtype=float)
    im = ax[1].imshow(P, cmap="viridis", aspect="auto", vmin=0, vmax=1); plt.colorbar(im, ax=ax[1], label="AP@50")
    ax[1].set_yticks(range(len(names))); ax[1].set_yticklabels([LABEL[n] for n in names], fontsize=6.5)
    ax[1].set_xticks(range(len(det["_classes"]))); ax[1].set_xticklabels(det["_classes"], rotation=45, fontsize=7); ax[1].grid(False)
    ax[1].set_title("(b) per-class AP@50")
    save_fig("FIG11_detection")
    CAP["FIG11_detection"] = "Downstream detection: one YOLO detector trained on HR-proxy images, applied to the test set as reconstructed by each method."


@_safe
def fig12_cv(ctx):
    p = DIRS["tables"] / "T4b_cv_per_fold.csv"
    if not p.exists():
        return
    d = pd.read_csv(p)
    fig, ax = plt.subplots(1, 2, figsize=(10, 3.6))
    for a, k in zip(ax, ("ssim", "psnr")):
        for m in d.method.unique():
            s = d[d.method == m].sort_values("fold"); a.plot(s.fold, s[k], "o-", color=COLOR[m], label=LABEL[m])
        a.set_xlabel("fold"); a.set_ylabel(MLAB[k]); a.set_xticks(range(C["CV_FOLDS"]))
    ax[0].legend(fontsize=7); ax[0].set_title("(a) SSIM per fold"); ax[1].set_title("(b) PSNR per fold")
    save_fig("FIG12_cross_validation")
    CAP["FIG12_cross_validation"] = "Session-grouped 5-fold cross-validation over the full dataset (seed 42)."


@_safe
def fig13_curves(ctx):
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
    for m in [m for m in ORDER if m in METHODS]:
        cs = [pd.read_csv(DIRS["logs"] / f"{m}_s{s}_curve.csv") for s in ctx["seeds_of"].get(m, []) if (DIRS["logs"] / f"{m}_s{s}_curve.csv").exists()]
        if not cs:
            continue
        L = min(len(c) for c in cs)
        for a, k in zip(ax, ("val_ssim", "val_psnr")):
            Mx = np.stack([c[k].values[:L] for c in cs]); it = cs[0]["iter"].values[:L]
            a.plot(it, Mx.mean(0), color=COLOR[m], label=LABEL[m])
            if len(cs) > 1:
                a.fill_between(it, Mx.min(0), Mx.max(0), color=COLOR[m], alpha=.12)
    for a in ax:
        a.axvline(C["GAN_WARMUP_ITERS"], ls=":", c="k", lw=.8); a.set_xlabel("iteration")
    ax[0].set_ylabel("validation SSIM"); ax[1].set_ylabel("validation PSNR"); ax[0].legend(fontsize=5.5, ncol=2)
    save_fig("FIG13_training_curves")
    CAP["FIG13_training_curves"] = "Validation curves (mean over seeds, band = seed range). Dotted line: end of the non-adversarial warm-up."


@_safe
def fig14_complexity(ctx):
    T1 = ctx["T1"]; col = f"{MLAB['ssim_fg']} mean" if f"{MLAB['ssim_fg']} mean" in T1 else f"{MLAB['ssim']} mean"
    plt.figure(figsize=(6, 4.2))
    for _, r in T1.iterrows():
        plt.scatter(max(r["Params (M)"], 1e-3), r[col], color=COLOR[r.key], s=40)
        plt.annotate(r.Method, (max(r["Params (M)"], 1e-3), r[col]), fontsize=6)
    plt.xscale("log"); plt.xlabel("parameters (M, log)"); plt.ylabel(col); plt.title("Quality vs model size")
    save_fig("FIG14_complexity")
    CAP["FIG14_complexity"] = "Reconstruction quality against model size."


@_safe
def fig15_native(ctx):
    w = DIRS["models"] / f"ours_s{C['SEEDS'][0]}_G_fp16.pt"
    if not w.exists():
        return
    G = build_generator("rrdb"); G.load_state_dict({k: v.float() for k, v in torch.load(w, map_location=DEVICE).items()}); G.eval()
    sub = INDEX[INDEX.split == "test"].sort_values("stem").iloc[:3]
    fig, ax = plt.subplots(len(sub), 3, figsize=(9, 3 * len(sub)), squeeze=False)
    with torch.no_grad():
        for i, (_, r) in enumerate(sub.iterrows()):
            x = to_t(get_img(r.path)).to(DEVICE)
            for j, (img, t) in enumerate(((x, "native image (input)"), (bicubic_up(x).clamp(0, 1), f"bicubic ×{C['SCALE']}"),
                                          (G(x), f"Ours ×{C['SCALE']} (no reference exists)"))):
                ax[i, j].imshow(img[0, 0].cpu().numpy(), cmap="gray", vmin=0, vmax=1); ax[i, j].axis("off")
                if i == 0:
                    ax[i, j].set_title(t, fontsize=8)
    del G
    free_gpu()
    save_fig("FIG15_native_inference")
    CAP["FIG15_native_inference"] = ("The model applied to native dataset images (no synthetic degradation). No higher-resolution reference exists, "
                                     "so this figure is qualitative only; quantitative claims refer to the HR proxy.")


@_safe
def fig17_bands(ctx):
    T8 = ctx.get("T8")
    if T8 is None or not len(T8):
        return
    fig, ax = plt.subplots(1, 2, figsize=(12, 3.9))
    x = np.arange(len(T8))
    ax[0].bar(x - .2, T8.rmse_observable, .4, label="observable band (< LR Nyquist)", color="#1f77b4")
    ax[0].bar(x + .2, T8.rmse_unobservable, .4, label="unobservable band (> LR Nyquist)", color="#d62728")
    ax[0].set_xticks(x); ax[0].set_xticklabels([LABEL[m] for m in T8.method], rotation=50, ha="right", fontsize=6.5)
    ax[0].set_ylabel("RMSE vs HR proxy"); ax[0].legend(fontsize=7); ax[0].set_title("(a) where the reconstruction error lies")
    ax[1].bar(x, T8.hf_energy_ratio_sr_over_hr, color=[COLOR[m] for m in T8.method])
    ax[1].axhline(1, ls="--", c="k", lw=.8)
    ax[1].set_xticks(x); ax[1].set_xticklabels([LABEL[m] for m in T8.method], rotation=50, ha="right", fontsize=6.5)
    ax[1].set_ylabel("HF energy (SR) / HF energy (HR proxy)"); ax[1].set_title("(b) synthesised vs missing high-frequency detail")
    save_fig("FIG17_observable_vs_unobservable")
    CAP["FIG17_observable_vs_unobservable"] = (
        "Error decomposition at the LR Nyquist frequency. A PSF / data-consistency term constrains only the observable band; "
        "detail above it (where adversarial models synthesise texture, ratio > 1) is invisible to the measurement, which explains "
        "why the PSF loss cannot regularise GAN hallucinations.")


def paper_figures(ctx):
    print(f"\n{'#' * 72}\nPAPER FIGURES\n{'#' * 72}")
    apply_style()
    for fn in (fig01_physics, fig02_qualitative, fig03_main, fig04_perception_distortion, fig05_foreground, fig06_per_image, fig07_ablation,
               fig08_data_consistency, fig09_spectrum, fig10_shift, fig11_detection, fig12_cv, fig13_curves, fig14_complexity, fig15_native,
               fig17_bands):
        fn(ctx)


# ==============================================================================
# NUMBERS FOR TEXT, EVIDENCE SUMMARY, ASSET INDEX
# ==============================================================================
def numbers_for_text(ctx):
    T1, T2, T3 = ctx["T1"], ctx["T2"], ctx["T3"]
    L = ["# Numbers for the abstract and text (generated from the result tables -- copy, do not retype)", "",
         f"Test set: {ctx['n_test']} images, session-grouped split; scale ×{C['SCALE']}; Gaussian PSF σ={C['PSF_SIGMA_U']}/{C['PSF_SIGMA_V']} px; "
         f"noise σ={C['NOISE_SIGMA']}; {C['ITERS']} iterations, batch {C['BATCH']}, {C['HR_CROP']}px HR crops.", ""]
    def g(key, k):
        r = T1[T1.key == key]
        return (r[f"{MLAB[k]} mean"].iloc[0], r[f"{MLAB[k]} sd"].iloc[0]) if len(r) and f"{MLAB[k]} mean" in r else (np.nan, np.nan)
    for key in [k for k in ORDER if k in set(T1.key)]:
        parts = []
        for k in ("ssim", "ssim_fg", "psnr", "lpips"):
            m, s = g(key, k)
            if not np.isnan(m):
                parts.append(f"{MLAB[k]} {m:.4f}" + (f" ± {s:.4f}" if not np.isnan(s) else ""))
        L.append(f"- {LABEL[key]}: " + "; ".join(parts))
    if len(T2):
        r = T2[(T2.comparison == f"{LABEL['ours']} vs {LABEL['bicubic']}")]
        for _, x in r.iterrows():
            L.append(f"- Ours vs bicubic, {x.metric}: mean diff {x.mean_diff:+.4f}, better on {x['win_rate_%']:.1f}% of images, "
                     f"Wilcoxon p={x.wilcoxon_p:.2e}, Holm p={x.p_holm:.2e}")
    if len(T3):
        for _, x in T3[T3.effect.str.startswith("PSF")].iterrows():
            tcol = "TOST_p"
            L.append(f"- {x.effect}, {x.metric}: Δ={x.mean_diff:+.4f} (n={x.n_seeds} seeds, 95% CI {x.t_ci95}), paired t p={x.paired_t_p:.3f}, "
                     f"TOST p={x[tcol]:.3f}")
    if ctx["det"]:
        for n in ("hr_proxy", "bicubic", "rrdb_l1", "ours_nopsf", "ours"):
            if n in ctx["det"]:
                d = ctx["det"][n]
                L.append(f"- Detection on {LABEL[n]}: mAP@50 {d['mAP50']:.3f} [{d['mAP50_ci'][0]:.3f}, {d['mAP50_ci'][1]:.3f}], mAP@50:95 {d['mAP50_95']:.3f}")
    if ctx["det"] and ctx["det"].get("_pairs"):
        for x in ctx["det"]["_pairs"]:
            L.append(f"- Detection {LABEL[x['A']]} − {LABEL[x['B']]}: ΔmAP@50 {x['delta_mAP50']:+.3f} "
                     f"[{x['ci95'][0]:.3f}, {x['ci95'][1]:.3f}], paired bootstrap p={x['p_boot']:.3f}")
    ext = jload(DIRS["analysis"] / "external_thz_tds.json")
    if ext:
        L.append(f"- External THz-TDS (real measurement, CAD ground truth, {ext['n_objects']} objects / {ext['n_views']} views, zero-shot):")
        for r_ in ext["table"]:
            L.append(f"    {r_['method']}: calibrated SSIM {r_.get('ssim_cal_mean', float('nan')):.4f}, calibrated SSIM-FG "
                     f"{r_.get('ssim_fg_cal_mean', float('nan')):.4f}, edge correlation {r_.get('grad_corr_mean', float('nan')):.4f} "
                     f"(raw SSIM {r_.get('ssim_mean', float('nan')):.4f}, raw PSNR {r_.get('psnr_mean', float('nan')):.2f} dB)")
        for k_, v_ in ext["tests"].items():
            L.append(f"    {k_}: Δ={v_['mean_diff']:+.4f}, better on {v_['objects_better']}/{v_['n_objects']} objects, Wilcoxon p={v_['wilcoxon_p']:.3f}")
    (R / "NUMBERS_FOR_TEXT.md").write_text("\n".join(L), encoding="utf-8")
    print("\n" + "\n".join(L))


def evidence_summary(ctx):
    T1, T3, det = ctx["T1"], ctx["T3"], ctx["det"]
    L = ["=" * 72, "PRELIMINARY -- NOT FOR THE PAPER" if ctx["prelim"] else "FINAL EVIDENCE SUMMARY (no claim is auto-asserted)", "=" * 72]

    def best_baseline(k):
        sub = T1[~T1.key.isin(["ours", "ours_nopsf", "ours_airy"])]
        col = f"{MLAB[k]} mean"
        return sub.loc[sub[col].idxmax() if HIGHER_BETTER[k] else sub[col].idxmin()] if col in sub else None
    for k in ("ssim_fg", "psnr", "lpips"):
        b = best_baseline(k); o = T1[T1.key == "ours"]
        if b is not None and len(o):
            ov = o[f"{MLAB[k]} mean"].iloc[0]
            better = (ov > b[f"{MLAB[k]} mean"]) == HIGHER_BETTER[k]
            L.append(f"{MLAB[k]}: Ours {ov:.4f} vs strongest other method ({b.Method}) {b[f'{MLAB[k]} mean']:.4f} -> "
                     f"{'Ours better' if better else 'Ours NOT better'} (mean over seeds)")
    if len(T3):
        tcol = "TOST_p"
        for _, x in T3[T3.effect.str.startswith("PSF") & T3.metric.isin(["SSIM", "SSIM-FG", "PSNR (dB)", "‖T(x̂)−y‖₁"])].iterrows():
            verdict = ("significant difference" if x.paired_t_p < .05 else
                       "equivalent within margin" if x[tcol] < .05 else "inconclusive (neither different nor shown equivalent)")
            L.append(f"{x.effect} [{x.metric}]: Δ={x.mean_diff:+.4f}, p={x.paired_t_p:.3f}, TOST p={x[tcol]:.3f} -> {verdict}")
    if det and "ours" in det and "bicubic" in det:
        d = det["ours"]
        L.append(f"Detection mAP@50: ours {d['mAP50']:.3f} vs bicubic {det['bicubic']['mAP50']:.3f}, Δ={d.get('delta_vs_bicubic', np.nan):+.3f} "
                 f"CI {d.get('delta_vs_bicubic_ci')} | HR proxy {det.get('hr_proxy', {}).get('mAP50', np.nan):.3f}")
    ext = jload(DIRS["analysis"] / "external_thz_tds.json")
    if ext and ext.get("tests"):
        for key in ("ours_vs_bicubic_ssim_fg_cal", "ours_vs_rrdb_l1_ssim_fg_cal", "ours_vs_ours_nopsf_ssim_fg_cal", "rrdb_l1_vs_rl_deconv_ssim_fg_cal",
                    "ours_vs_bicubic_grad_corr", "rrdb_l1_vs_bicubic_grad_corr"):
            if key in ext["tests"]:
                t_ = ext["tests"][key]
                L.append(f"External THz-TDS {key}: Δ={t_['mean_diff']:+.4f}, better on {t_['objects_better']}/{t_['n_objects']} objects, "
                         f"Wilcoxon p={t_['wilcoxon_p']:.3f} (minimum possible p with n={t_['n_objects']}: {2 / 2 ** t_['n_objects']:.3f})")
    L += ["", "How to frame the paper:",
          "  - Main claim only what T1/T2/T6 support against the STRONGEST baseline, not only against bicubic.",
          "  - PSF loss: report as an empirical study; use the TOST verdict ('equivalent' / 'inconclusive'), never 'improves' unless p<0.05.",
          "  - Lead with foreground SSIM / LPIPS / detection; full-image SSIM is inflated by background denoising (FIG05).",
          "  - 'HR proxy/reference' everywhere; FIG15 is qualitative; PSF is calibrated, not measured; one dataset and one system.",
          "  - Copy every number from NUMBERS_FOR_TEXT.md and the T*.csv tables."]
    txt = "\n".join(L)
    (R / ("PRELIMINARY_EVIDENCE_SUMMARY.txt" if ctx["prelim"] else "FINAL_EVIDENCE_SUMMARY.txt")).write_text(txt, encoding="utf-8")
    print("\n" + txt)


def assets_index():
    figs = sorted({p.stem for p in DIRS["figures"].glob("*.pdf")}); tabs = sorted({p.stem for p in DIRS["tables"].glob("*.csv")})
    L = ["# Paper assets", "", "## Figures"] + [f"- **{f}** — {CAP.get(f, '')}" for f in figs] + ["", "## Tables"] + [f"- {t}" for t in tabs]
    (R / "PAPER_ASSETS.md").write_text("\n".join(L), encoding="utf-8")
    print(f"[assets] {len(figs)} figures, {len(tabs)} tables -> {R / 'PAPER_ASSETS.md'}")




# ==============================================================================
# 0. DATASET DOWNLOAD / PREPARATION
# ==============================================================================
DL = {
    "LINGLIX_GDRIVE_FOLDER": "1A6LiyWAvRmKIJN5yXQZ3HxZVwNEFz8uV",          # official Google Drive folder (README of LingLIx/THz_Dataset)
    "LINGLIX_EXPECTED_IMAGES": 3157,
    "MIN_IMAGES": 1000,                     # an attached copy must have at least this many images
    "TDS_GDRIVE_IDS": ["1fVY5sUPltpD_bsOt1np9bcDUnhsDTxsc", "1FqxHWqjZOZme_pzE7Nx9osHlDV4FOyXb"],   # both ids appear in the wtnthu README
    "TMP": "/kaggle/tmp" if os.path.isdir("/kaggle") else "/tmp/thz_tmp",
    "USE_TDS": True,
}


def _gdown():
    if importlib.util.find_spec("gdown") is None:
        _pip("gdown")
    import gdown
    return gdown


def extract_archives(folder):
    """Extract every .zip/.tar(.gz)/.rar/.7z found below `folder` (recursively, until nothing new)."""
    import zipfile, tarfile
    done = set()
    for _ in range(3):
        new = False
        for p in sorted(Path(folder).rglob("*")):
            if p in done or not p.is_file():
                continue
            low = p.name.lower(); out = p.parent / (p.stem + "_x")
            try:
                if low.endswith(".zip"):
                    with zipfile.ZipFile(p) as z:
                        z.extractall(out)
                elif low.endswith((".tar", ".tar.gz", ".tgz")):
                    with tarfile.open(p) as t:
                        t.extractall(out)
                elif low.endswith((".rar", ".7z")):
                    if shutil.which("7z") is None:
                        subprocess.run("apt-get -qq update && apt-get -qq install -y p7zip-full p7zip-rar >/dev/null 2>&1", shell=True)
                    if shutil.which("7z"):
                        subprocess.run(["7z", "x", "-y", f"-o{out}", str(p)], stdout=subprocess.DEVNULL)
                    elif shutil.which("unrar"):
                        out.mkdir(exist_ok=True); subprocess.run(["unrar", "x", "-o+", str(p), str(out)], stdout=subprocess.DEVNULL)
                    else:
                        print(f"[download] cannot extract {p.name}: no 7z/unrar available"); done.add(p); continue
                else:
                    continue
                done.add(p); new = True
                print(f"[download] extracted {p.name}")
            except Exception as e:
                print(f"[download] extraction failed for {p.name}: {e}"); done.add(p)
        if not new:
            break


def _copy_tree_images(src_root, dst, files):
    t0 = time.time()
    for k, p in enumerate(files, 1):
        rel = Path(p).relative_to(src_root)
        q = dst / rel
        q.parent.mkdir(parents=True, exist_ok=True)
        if not q.exists():
            shutil.copyfile(p, q)
        if k % 500 == 0 or k == len(files):
            el = time.time() - t0
            print(f"[dataset] copied {k}/{len(files)} files ({el:.0f}s, about {(len(files) - k) * el / k / 60:.1f} min left)", flush=True)


def prepare_linglix():
    """Returns a folder holding the active-THz images + VOC xml, stored inside the results directory
    so that image paths stay valid in every later session."""
    dst = R / "datasets" / "linglix"
    have = [p for p in dst.rglob("*") if p.suffix.lower() in IMG_EXT] if dst.exists() else []
    if len(have) >= DL["MIN_IMAGES"]:
        print(f"[dataset] active-THz dataset ready: {len(have)} images in {dst}")
        return str(dst)
    dst.mkdir(parents=True, exist_ok=True)
    # (1) attached Kaggle input
    print("[dataset] scanning /kaggle/input for an attached copy (can take a few minutes on large inputs) ...", flush=True)
    t_scan = time.time()
    allp = [p for p in glob.glob("/kaggle/input/**/*", recursive=True)
            if p.lower().endswith(IMG_EXT + (".xml",)) and "/thz_sr_q1/" not in p and "/Gt/" not in p]
    print(f"[dataset] scan finished in {time.time() - t_scan:.0f}s: {len(allp)} image/xml files found", flush=True)
    if allp:
        key = Counter(_ds_key(p) for p in allp if p.lower().endswith(IMG_EXT)).most_common(1)
        if key and key[0][1] >= DL["MIN_IMAGES"]:
            root = key[0][0]
            files = [p for p in allp if _ds_key(p) == root]
            print(f"[dataset] copying attached dataset {root} ({key[0][1]} images) ...")
            _copy_tree_images(root, dst, files)
    # (2) authors' Google Drive folder
    have = [p for p in dst.rglob("*") if p.suffix.lower() in IMG_EXT]
    if len(have) < DL["MIN_IMAGES"]:
        tmp = Path(DL["TMP"]) / "linglix"; tmp.mkdir(parents=True, exist_ok=True)
        print("[dataset] downloading the active-THz dataset from the authors' Google Drive folder ...")
        try:
            _gdown().download_folder(id=DL["LINGLIX_GDRIVE_FOLDER"], output=str(tmp), quiet=False, remaining_ok=True)
        except Exception as e:
            print(f"[dataset] Google Drive download failed: {e}")
        extract_archives(tmp)
        files = [str(p) for p in tmp.rglob("*") if p.suffix.lower() in IMG_EXT + (".xml",)]
        _copy_tree_images(tmp, dst, files)
        have = [p for p in dst.rglob("*") if p.suffix.lower() in IMG_EXT]
    if len(have) < DL["MIN_IMAGES"]:
        raise RuntimeError("Active-THz dataset not available. Download THZ_dataset_det_VOC from "
                           "https://github.com/LingLIx/THz_Dataset (Google Drive link), upload it as a Kaggle dataset, attach it, re-run.")
    n_xml = len(list(dst.rglob("*.xml")))
    print(f"[dataset] active-THz dataset ready: {len(have)} images, {n_xml} xml "
          f"(official release: {DL['LINGLIX_EXPECTED_IMAGES']} images){'' if len(have) == DL['LINGLIX_EXPECTED_IMAGES'] else '  <-- CHECK: count differs'}")
    return str(dst)


def sniff_archive(p):
    with open(p, "rb") as f:
        h = f.read(8)
    if h[:2] == b"PK":
        return "zip"
    if h[:4] == b"Rar!":
        return "rar"
    if h[:6] == b"7z\xbc\xaf\x27\x1c":
        return "7z"
    if h[:2] == b"\x1f\x8b" or h[:3] == b"BZh" or h[:6] == b"\xfd7zXZ\x00":
        return "tar"
    import tarfile
    try:
        if tarfile.is_tarfile(p):
            return "tar"
    except Exception:
        pass
    return "html" if h[:1] == b"<" else "unknown"


def _gt_time_dirs(root):
    """Case-insensitive: a 'gt' folder whose parent also holds a 'Time' folder (layout of the THz-TDS release)."""
    out = []
    for g in Path(root).rglob("*"):
        if g.is_dir() and g.name.lower() == "gt":
            t = [d for d in g.parent.iterdir() if d.is_dir() and d.name.lower() == "time"]
            if t:
                out.append((g, t[0]))
    return out


def _have_dirs(out):
    return _gt_time_dirs(out)


def get_7zz():
    """Official 7-Zip (reads RAR4/RAR5, 7z, zip, tar); downloaded once if not installed."""
    for name in ("7zz", "7zzs"):
        if shutil.which(name):
            return shutil.which(name)
    d = Path(DL["TMP"]) / "7zip"; exe = d / "7zz"
    if exe.exists():
        return str(exe)
    d.mkdir(parents=True, exist_ok=True)
    import urllib.request, tarfile
    for url in ("https://github.com/ip7z/7zip/releases/download/24.08/7z2408-linux-x64.tar.xz",
                "https://www.7-zip.org/a/7z2408-linux-x64.tar.xz"):
        try:
            urllib.request.urlretrieve(url, d / "7z.tar.xz")
            with tarfile.open(d / "7z.tar.xz") as t:
                t.extractall(d)
            if exe.exists():
                exe.chmod(0o755); print(f"[dataset] official 7-Zip installed ({url.split('/')[2]})", flush=True)
                return str(exe)
        except Exception as e:
            print(f"[dataset] 7-Zip download failed from {url.split('/')[2]}: {e}")
    return None


def _wanted(path):
    parts = [x.lower() for x in re.split(r"[\\/]", path)]
    return ("gt" in parts or "time" in parts) and path.lower().endswith(".npy")


def extract_gt_time(arc, out):
    """Extract only the gt/ and Time/ .npy files of the THz-TDS archive (zip, tar, rar4/5 or 7z)."""
    import zipfile, tarfile
    kind = sniff_archive(arc); out = Path(out); out.mkdir(parents=True, exist_ok=True)
    print(f"[dataset] archive type detected: {kind} ({os.path.getsize(arc) / 1e9:.2f} GB)", flush=True)
    try:
        if kind == "zip":
            with zipfile.ZipFile(arc) as z:
                z.extractall(out, members=[m for m in z.namelist() if _wanted(m)])
        elif kind == "tar":
            with tarfile.open(arc, "r:*") as t:
                t.extractall(out, members=[m for m in t.getmembers() if _wanted(m.name)])
        elif kind in ("rar", "7z"):
            exe = get_7zz()
            if exe:
                r = subprocess.run([exe, "l", "-ba", "-slt", str(arc)], capture_output=True, text=True)
                paths = [l[7:] for l in r.stdout.splitlines() if l.startswith("Path = ")]
                want = [x for x in paths if _wanted(x)]
                print(f"[dataset] archive lists {len(paths)} entries; extracting {len(want)} gt/Time files ...", flush=True)
                lst = out / "_wanted.txt"
                lst.write_text("\n".join(want), encoding="utf-8")
                r = subprocess.run([exe, "x", "-y", f"-o{out}", "-scsUTF-8", str(arc), f"@{lst}"], capture_output=True, text=True)
                if r.returncode not in (0, 1):
                    print(f"[dataset] 7-Zip returned {r.returncode}: {(r.stderr or r.stdout)[-400:]}")
            if not _have_dirs(out):
                if shutil.which("unrar") is None:
                    subprocess.run("apt-get -qq update >/dev/null 2>&1; apt-get -qq install -y unrar >/dev/null 2>&1", shell=True)
                if shutil.which("unrar"):
                    print("[dataset] trying unrar ...", flush=True)
                    subprocess.run(["unrar", "x", "-o+", str(arc), str(out) + "/"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        else:
            print(f"[dataset] {Path(arc).name} is not an archive ({kind}); Google Drive may have returned a web page instead of the file")
    except Exception as e:
        print(f"[dataset] extraction error: {type(e).__name__}: {e}")
    found = _have_dirs(out)
    print(f"[dataset] gt/Time folder pairs found: {len(found)}", flush=True)
    return found


def _time_max(t, gt_shape):
    """Time-domain THz signal -> 'time-max' image (peak absolute field per pixel)."""
    t = np.asarray(t)
    if t.ndim == 2:
        return t.astype(np.float32)
    ax = [i for i in range(t.ndim) if t.shape[i] not in gt_shape]
    ax = ax[-1] if ax else int(np.argmax(t.shape))
    return np.abs(t).max(axis=ax).astype(np.float32)


def _norm01(a):
    lo, hi = np.percentile(a, 0.5), np.percentile(a, 99.5)
    return np.clip((a - lo) / max(hi - lo, 1e-8), 0, 1)


def prepare_thz_tds():
    """Real THz-TDS measurements + CAD ground truth -> list of (object, view, measured_u8, gt_u8), cached."""
    if not DL["USE_TDS"]:
        return None
    cache = R / "datasets" / "thz_tds_pairs.pkl"
    if cache.exists():
        import pickle
        pairs = pickle.load(open(cache, "rb"))
        print(f"[dataset] THz-TDS external set ready: {len(pairs)} views of {len({p[0] for p in pairs})} objects")
        return pairs
    gts = []
    for base in glob.glob("/kaggle/input/*/") + glob.glob("/kaggle/input/*/*/*/"):
        if "/thz_sr_q1/" not in base:
            gts += _gt_time_dirs(base) if os.path.isdir(base) and len(base) > 0 else []
        if gts:
            break
    if not gts:
        tmp = Path(DL["TMP"]) / "thz_tds"; tmp.mkdir(parents=True, exist_ok=True)
        if not list(tmp.rglob("Gt")):
            for fid in DL["TDS_GDRIVE_IDS"]:
                try:
                    print(f"[dataset] downloading THz-TDS dataset (Google Drive id {fid}) ...")
                    out = _gdown().download(id=fid, output=str(tmp / f"thz_tds_{fid[:6]}.zip"), quiet=False)
                    if out and os.path.getsize(out) > 1e6:
                        break
                except Exception as e:
                    print(f"[dataset] download failed for {fid}: {e}")
            for z in sorted(tmp.glob("thz_tds_*")):
                if z.is_file() and not _have_dirs(tmp / "x"):
                    extract_gt_time(z, tmp / "x")
        gts = _gt_time_dirs(tmp)
    if not gts:
        print("[dataset] THz-TDS dataset unavailable -> external evaluation skipped "
              "(download from github.com/wtnthu/THz_Data, upload to Kaggle, attach, re-run).")
        return None
    gt_dir, tm_dir = gts[0]
    pairs, s, skipped = [], C["SCALE"], 0
    for obj_dir in sorted(d.name for d in gt_dir.iterdir() if d.is_dir()):
        if not (tm_dir / obj_dir).is_dir():
            print(f"[dataset] {obj_dir}: no Time folder -> skipped"); continue
        obj = re.sub(r"^td_freq_", "", obj_dir)
        for f in sorted((gt_dir / obj_dir).glob("*.npy"), key=lambda x: int(re.sub(r"\D", "", x.stem) or 0)):
            view = re.sub(r"^gt_", "", f.stem)
            if not re.fullmatch(r"\d+", view):
                skipped += 1; continue
            tf = tm_dir / obj_dir / f"{view}.npy"
            if not tf.exists():
                skipped += 1; continue
            g = np.asarray(np.load(f), dtype=np.float32).squeeze()
            m = _time_max(np.load(tf, mmap_mode="r"), g.shape)
            if m.shape != g.shape:
                m = np.asarray(Image.fromarray(m).resize((g.shape[1], g.shape[0]), Image.BILINEAR), dtype=np.float32)
            g, m = _norm01(g), _norm01(m)
            if np.corrcoef(g.ravel(), m.ravel())[0, 1] < 0:          # transmission: object attenuates -> invert
                m = 1 - m
            H, W = g.shape[0] - g.shape[0] % s, g.shape[1] - g.shape[1] % s
            if min(H, W) < 4 * s * 4:
                continue
            pairs.append((obj, view, (m[:H, :W] * 255).round().astype(np.uint8), (g[:H, :W] * 255).round().astype(np.uint8)))
    import pickle
    cache.parent.mkdir(parents=True, exist_ok=True)
    pickle.dump(pairs, open(cache, "wb"))
    print(f"[dataset] THz-TDS external set prepared: {len(pairs)} views, objects {sorted({p[0] for p in pairs})}")
    return pairs


# ==============================================================================
# EXTERNAL EVALUATION: real THz-TDS measurements vs CAD ground truth (zero-shot)
# ==============================================================================
def load_saved_generator(method):
    w = DIRS["models"] / f"{method}_s{C['SEEDS'][0]}_G_fp16.pt"
    if not w.exists():
        return None
    G = build_generator(METHODS[method]["gen"])
    G.load_state_dict({k: v.float() for k, v in torch.load(w, map_location=DEVICE).items()}); G.eval()
    return G


@torch.no_grad()
def external_eval(pairs):
    out_p = DIRS["analysis"] / "external_thz_tds.json"
    old = jload(out_p)
    if pairs is None or (old and old.get("protocol_version") == 2):
        return old
    print(f"\n{'#' * 72}\nEXTERNAL TEST: THz-TDS (real measurement, CAD ground truth), zero-shot\n{'#' * 72}")
    systems = {"bicubic": CLASSICAL_FN["bicubic"], "rl_deconv": CLASSICAL_FN["rl_deconv"]}
    for m in C["SAVE_G_FOR"]:
        G = load_saved_generator(m)
        if G is not None:
            systems[m] = (lambda G_: (lambda lr: G_(lr)))(G)
    from sklearn.isotonic import IsotonicRegression
    from scipy.ndimage import sobel

    def calibrate(pred, gt_, seed):
        """Per-image monotonic intensity mapping onto the CAD scale (direction chosen automatically); fitted on a
        fixed random pixel subset. Removes the non-linear transmission-vs-thickness mismatch between modalities."""
        p_, g_ = pred.ravel(), gt_.ravel()
        idx = np.random.default_rng(seed).choice(p_.size, min(20000, p_.size), replace=False)
        ir = IsotonicRegression(y_min=0.0, y_max=1.0, increasing="auto", out_of_bounds="clip").fit(p_[idx], g_[idx])
        return ir.predict(p_).reshape(pred.shape).astype(np.float32)

    def grad_corr(a, b):
        ga = np.hypot(sobel(a, 0), sobel(a, 1)).ravel(); gb = np.hypot(sobel(b, 0), sobel(b, 1)).ravel()
        return float(np.corrcoef(ga, gb)[0, 1]) if ga.std() > 0 and gb.std() > 0 else np.nan

    rows, qual = [], {}
    for obj, view, meas, gt in pairs:
        x_meas = to_t(meas).to(DEVICE); x_gt = to_t(gt).to(DEVICE)
        lr = F.avg_pool2d(x_meas, C["SCALE"])                          # real THz blur; only the pixel down-sampling is synthetic
        mask = gt > 0.05 * 255
        for name, fn in {"measured_fullres": None, **systems}.items():
            sr = x_meas if fn is None else fn(lr)
            mse = F.mse_loss(sr, x_gt).item(); sm = ssim_map(sr, x_gt)[0, 0]
            r = {"object": obj, "view": view, "method": name, "psnr": 10 * math.log10(1 / (mse + 1e-10)), "ssim": sm.mean().item()}
            if 0.01 < mask.mean() < 0.99:
                r["ssim_fg"] = sm[torch.from_numpy(mask[5:-5, 5:-5]).to(DEVICE)].mean().item()
            srn, gtn = sr[0, 0].float().cpu().numpy(), gt.astype(np.float32) / 255.0
            cal = calibrate(srn, gtn, stable_int(obj, view))
            ct = torch.from_numpy(cal)[None, None].to(DEVICE)
            smc = ssim_map(ct, x_gt)[0, 0]
            r["psnr_cal"] = 10 * math.log10(1 / (F.mse_loss(ct, x_gt).item() + 1e-10)); r["ssim_cal"] = smc.mean().item()
            if 0.01 < mask.mean() < 0.99:
                r["ssim_fg_cal"] = smc[torch.from_numpy(mask[5:-5, 5:-5]).to(DEVICE)].mean().item()
            r["grad_corr"] = grad_corr(srn, gtn)
            lp = lpips_fn()
            if lp:
                r["lpips"] = lp(sr.repeat(1, 3, 1, 1) * 2 - 1, x_gt.repeat(1, 3, 1, 1) * 2 - 1).item()
            rows.append(r)
            if view == "0":
                qual.setdefault(obj, {})[name] = (sr[0, 0].cpu().numpy() * 255).astype(np.uint8)
        if view == "0":
            qual[obj]["gt"] = gt
    df = pd.DataFrame(rows)
    df.to_csv(DIRS["tables"] / "X0_external_thz_tds_per_view.csv", index=False)
    per_obj = df.groupby(["method", "object"]).mean(numeric_only=True).reset_index()       # unit of analysis = object (n=7)
    order = ["measured_fullres", "bicubic", "rl_deconv"] + [m for m in ORDER if m in systems and m not in CLASSICAL]
    KX = ("psnr_cal", "ssim_cal", "ssim_fg_cal", "grad_corr", "psnr", "ssim", "ssim_fg", "lpips")
    X1 = per_obj.groupby("method").agg(objects=("object", "nunique"), **{f"{k}_mean": (k, "mean") for k in KX if k in per_obj},
                                       **{f"{k}_sd": (k, "std") for k in ("psnr_cal", "ssim_cal", "ssim_fg_cal", "grad_corr") if k in per_obj}).reindex(
        [o for o in order if o in set(per_obj.method)]).reset_index()
    X1["method"] = X1.method.map(lambda m: "Measured image (full-res, no SR)" if m == "measured_fullres" else LABEL[m])
    save_table(X1.round(4), "X1_external_thz_tds",
               "Zero-shot evaluation on real THz-TDS measurements with CAD ground truth (input = measured image down-sampled ×s), "
               "mean ± SD over objects. *_cal: after a per-image monotonic intensity calibration onto the CAD scale (identical for "
               "every method, including the measured image); grad_corr: calibration-free edge agreement. Raw PSNR/SSIM are "
               "dominated by the transmission-vs-thickness intensity mismatch and are reported for transparency only.")
    print(X1.round(4).to_string(index=False))
    tests = {}
    for A_, refs in (("ours", ("bicubic", "rl_deconv", "rrdb_l1", "ours_nopsf")), ("rrdb_l1", ("bicubic", "rl_deconv"))):
        if A_ not in systems:
            continue
        for ref in refs:
            if ref not in systems:
                continue
            for k in ("ssim_cal", "ssim_fg_cal", "grad_corr"):
                if k not in per_obj:
                    continue
                a = per_obj[per_obj.method == A_].set_index("object")[k]
                b = per_obj[per_obj.method == ref].set_index("object")[k].reindex(a.index)
                try:
                    p = float(sps.wilcoxon(a.values, b.values).pvalue)
                except Exception:
                    p = np.nan
                tests[f"{A_}_vs_{ref}_{k}"] = {"mean_diff": float((a - b).mean()), "objects_better": int((a > b).sum()),
                                               "n_objects": int(len(a)), "wilcoxon_p": p}
    apply_style()
    objs = sorted(qual)[:4]
    cols = [c for c in ["bicubic", "rl_deconv", "rrdb_l1", "ours_nopsf", "ours"] if c in systems]
    if objs:
        fig, ax = plt.subplots(len(objs), len(cols) + 2, figsize=(2.1 * (len(cols) + 2), 2.0 * len(objs)), squeeze=False)
        for i, o in enumerate(objs):
            ims = [("measured (input before ↓s)", qual[o]["measured_fullres"])] + [(LABEL[c], qual[o][c]) for c in cols] + [("CAD ground truth", qual[o]["gt"])]
            for j, (t, im) in enumerate(ims):
                ax[i, j].imshow(im, cmap="gray", vmin=0, vmax=255); ax[i, j].axis("off")
                if i == 0:
                    ax[i, j].set_title(t, fontsize=7)
            ax[i, 0].text(-5, im.shape[0] / 2, o, rotation=90, va="center", ha="right", fontsize=8)
        save_fig("FIG16_external_thz_tds")
        CAP["FIG16_external_thz_tds"] = ("Zero-shot test on real THz-TDS transmission measurements (Hung & Yang) with ground truth from the "
                                         "3D-printing models; models were trained only on the active-THz dataset.")
    res = {"protocol_version": 2, "table": X1.to_dict("records"), "tests": tests, "n_views": int(len(pairs)), "n_objects": int(df.object.nunique()),
           "protocol": "input = measured time-max image average-pooled by the SR scale; output compared with the CAD projection after a "
                       "per-image monotonic intensity calibration (raw metrics kept); statistics over objects (views are not independent; "
                       "30 of 60 views per object are horizontal flips)"}
    jdump(res, out_p)
    return res


# ==============================================================================
# MASTER DRIVER (single cell, resumable)
# ==============================================================================
apply_style()
stage("START: restoring previous results (if attached)")
restore_previous()
jdump(CFG, R / "config_used.json")
print("[env]", env_info())
NO_GPU = not torch.cuda.is_available() and not ALLOW_CPU
if NO_GPU:
    print("\n" + "!" * 72 + "\nNO GPU IN THIS SESSION (torch reports CPU only) -> NO TRAINING.\n"
          + ("CPU mode: building a PRELIMINARY analysis from the finished runs (tables, statistics, figures,\n"
             "external THz-TDS test). Detection and the remaining runs need a GPU session later.\n"
             if CPU_PRELIMINARY_ANALYSIS else "Set Accelerator -> GPU T4 x1 and run again.\n")
          + "Completed runs are never modified.\n" + "!" * 72, flush=True)
    if not CPU_PRELIMINARY_ANALYSIS:
        print("No GPU and CPU_PRELIMINARY_ANALYSIS=False -> nothing to do in this session.")
stage("STEP 0a: active-THz dataset")
C["DATA_ROOT"] = prepare_linglix()
stage("STEP 0b: THz-TDS external dataset")
TDS_PAIRS = prepare_thz_tds()
stage("STEP 1: session-grouped split + leakage audit")
INDEX = stage_data()
OP = make_op("gaussian")
VAL = val_set(INDEX, C["VAL_N"])
supersede_warmup_gan_runs()
if NO_GPU:
    stage("CPU MODE: external THz-TDS test + PRELIMINARY final analysis (no training)")
    _have = all(run_done(f"{m}_s{C['SEEDS'][0]}") for m in C["SAVE_G_FOR"])
    if _have:
        try:
            external_eval(TDS_PAIRS)
        except Exception as _e:
            print(f"[external] failed: {type(_e).__name__}: {_e}")
    else:
        print("[external] skipped: not all seed-42 models in SAVE_G_FOR are finished yet")
    FORCE_FINAL = True
    CTX = final_analysis()
    _left = [r for r in build_queue() if not run_done(r)]
    print(f"\nPRELIMINARY analysis written (files named PRELIMINARY_* / tables + figures in {R}).\n"
          f"{len(_left)} main runs still need a GPU: {_left}\n"
          "Do not change any setting based on these numbers; the final analysis is regenerated after the last GPU session.")


def _gpu_pipeline():
    global CTX
    stage("STEP 2: classical baselines (bicubic, Richardson-Lucy)")
    for _m in CLASSICAL:
        execute_classical(_m)

    stage("STEP 3: training queue (prints every 250 iterations)")
    run_queue(C["RUNS_PER_SESSION"])
    queue_status()
    progress_report()
    try:
        ablation_preview()
    except Exception as _e:
        print("[preview] skipped:", _e)

    if all(run_done(r) for r in build_queue()):
        stage("STEP 4: downstream detection")
        if DO_DETECTION:
            try:
                stage_detection()
            except Exception as _e:
                print(f"[det] failed: {type(_e).__name__}: {_e}")
        stage("STEP 5: 5-fold cross-validation")
        if DO_CV:
            C["SESSION_HOURS"] -= FINAL_RESERVE_H            # keep time for the final analysis
            run_cv()
            C["SESSION_HOURS"] += FINAL_RESERVE_H
        stage("STEP 6: external THz-TDS test")
        try:
            external_eval(TDS_PAIRS)
        except Exception as _e:
            print(f"[external] failed: {type(_e).__name__}: {_e}")
        stage("STEP 7: final statistics, tables, figures")
        CTX = final_analysis()
        cv_done = all(run_done(r) for r in cv_queue())
        det_done = (DIRS["det"] / "det_results.json").exists() or not DO_DETECTION
        print("\nALL STAGES COMPLETE." if (cv_done and det_done) else
              f"\nMain queue complete; remaining: {'CV ' if not cv_done else ''}{'detection' if not det_done else ''} -> run this cell again next session "
              f"(the final tables/figures are regenerated each time).")
    else:
        left = [r for r in build_queue() if not run_done(r)]
        print(f"\n{len(left)} main runs left -> save this version, attach its output next session and run this cell again.")
    print(f"Session time left {hours_left():.1f}h. Results in {R}")


if NO_GPU:
    print('CPU session finished normally (no training).')
else:
    _gpu_pipeline()
```
